### Writing Proofs in Linear Algebra

This is Problem Set 1.P4, the second of the two honors options for Chapter One. Where Problem Set 1.P3, __Linear Systems with MATLAB__, asked you to compute, this one asks you to prove. Nothing here requires a computer.

Everything you have done in Chapter One rests on properties of the real numbers that were never stated: that you may add equations, that you may multiply a row by a nonzero number and divide it back, that a pivot can be scaled to $1$. Section 1.P4.1 states those properties as axioms and adds the one axiom that separates the real numbers from the rational numbers. Sections 1.P4.3 through 1.P4.5 then prove, carefully, the facts about linear combinations, spans, and solution sets that Chapter One used informally.

<div class="alert alert-block alert-success">

#### A Note to the Student

A proof is not a longer explanation. It is an argument in which every step follows from a definition, an axiom, or a previously proved statement, and in which a reader who disagrees with you has nowhere to stand. Learning to write one is the point of this problem set.

The exercises in section 1.P4.7 are independent of one another and are labeled by topic. Your instructor will assign a subset. Write your solutions in complete sentences. A chain of equations with no words is not a proof, and neither is a correct calculation with the quantifiers left out. When a claim is false you must exhibit a counterexample; when it is true you must prove it for every case the statement covers, and no number of examples will do.
</div>

### What We Have Been Assuming

Every elementary row operation in Chapter One used arithmetic in $\mathbb{R}$ without comment. Adding a multiple of one row to another uses distributivity; scaling a row by a nonzero number and later undoing it uses multiplicative inverses; reordering terms in an equation uses commutativity. These properties are not theorems about the real numbers. They are the axioms that define them.

<div class="alert alert-block alert-info">

#### <a style="text-decoration: none" class="anchor" id="Def:P4Field">Definition</a>

__The Field Axioms__

A __field__ is a set $F$ with two operations, addition and multiplication, satisfying the following for all $a$, $b$, $c$ in $F$.

$$
\begin{array}{|l|l|l|}
    \hline
    & \textbf{Addition} & \textbf{Multiplication} \\
    \hline
    \text{Associative} & (a+b)+c = a+(b+c) & (ab)c = a(bc) \\
    \hline
    \text{Commutative} & a+b = b+a & ab = ba \\
    \hline
    \text{Identity} & a + 0 = a & a\cdot 1 = a \\
    \hline
    \text{Inverse} & a + (-a) = 0 & a\,a^{-1} = 1 \ \ \text{for } a \neq 0 \\
    \hline
    \text{Distributive} & a(b+c) = ab + ac & (a+b)c = ac + bc \\
    \hline
\end{array}
$$

with $0 \neq 1$. Both $\mathbb{R}$ and $\mathbb{C}$ are fields, and they are the only fields used in this course.
</div>

The set $\mathbb{Z}$ of integers satisfies every one of these axioms except the last one in the multiplication column: the integer $2$ has no integer multiplicative inverse. That single failure is why Gaussian elimination over the integers cannot always scale a pivot to $1$.

There exist fields with only finitely many elements, and you will meet the two-element field in coding theory, cryptography, and hashing if you work in those areas. We will not use them in this course.

<div class="alert alert-block alert-info">

#### <a style="text-decoration: none" class="anchor" id="Def:P4Order">Definition</a>

__The Order Axioms__

The real numbers carry a relation $<$ such that for all $a$, $b$, $c$ in $\mathbb{R}$:

1. exactly one of $a < b$, $a = b$, $b < a$ holds;
2. if $a < b$ and $b < c$ then $a < c$;
3. if $a < b$ then $a + c < b + c$;
4. if $a < b$ and $0 < c$ then $ac < bc$.

A field with such a relation is an __ordered field__. Both $\mathbb{Q}$ and $\mathbb{R}$ are ordered fields.
</div>

The complex numbers form a field but __not__ an ordered field. No relation $<$ on $\mathbb{C}$ satisfies these four axioms, because $i \neq 0$ forces $i^2 > 0$, and $i^2 = -1$. This is worth remembering: every statement about a complex quantity being positive, or smallest, or bounded, must be a statement about its modulus $|z|$, which is a real number. Later work with complex vectors depends on that distinction.

### The Axiom That Separates $\mathbb{Q}$ from $\mathbb{R}$

The rational numbers satisfy every axiom stated so far. One more axiom is needed, and it is the reason the real numbers exist at all.

<div class="alert alert-block alert-info">

#### <a style="text-decoration: none" class="anchor" id="Def:P4Completeness">Definition</a>

__Bounds and the Completeness Axiom__

Let $S \subseteq \mathbb{R}$ be nonempty. A number $M$ is an __upper bound__ for $S$ if $x \leq M$ for every $x \in S$, and $S$ is __bounded above__ if it has an upper bound. A number $L$ is a __least upper bound__, or __supremum__, of $S$ if $L$ is an upper bound for $S$ and $L \leq M$ for every upper bound $M$ of $S$.

__Completeness Axiom.__ Every nonempty subset of $\mathbb{R}$ that is bounded above has a least upper bound in $\mathbb{R}$.
</div>

The rational numbers fail this axiom. The set

$$
S = \left\{\, x \in \mathbb{Q} \,:\, x > 0 \ \text{ and } \ x^2 < 2 \,\right\}
$$

is nonempty and bounded above by $2$, yet no rational number is a least upper bound for it; Exercise 2 asks you to prove this. Informally, $\mathbb{Q}$ has a hole where $\sqrt{2}$ should be. The real numbers are what you get when every such hole is filled, and one consequence is that every real number is a limit of a sequence of rational numbers. The decimal expansion $1,\ 1.4,\ 1.41,\ 1.414,\ \ldots$ is exactly such a sequence, made of rational numbers, converging to a real number that is not rational.

Completeness is not a technicality you can postpone. Chapter Two computes powers $P^k$ of a Markov matrix and asks what happens as $k$ grows; that question has no meaning without limits, and limits have no theory without completeness.

<div class="alert alert-block alert-info">

#### <a style="text-decoration: none" class="anchor" id="Def:P4Convergence">Definition</a>

__Convergence of a Sequence of Real Numbers__

A sequence $(a_n)$ of real numbers __converges__ to $L \in \mathbb{R}$, written

$$
\lim_{n \to \infty} a_n = L
$$

if for every $\varepsilon > 0$ there exists a positive integer $N$ such that

$$
|a_n - L| < \varepsilon \qquad \text{for every } n > N
$$

A sequence that converges to no such $L$ __diverges__.
</div>

Read the definition in the order it is written. The adversary chooses $\varepsilon$ first, and $\varepsilon$ may be as small as they like. You must then produce an $N$, and your $N$ is allowed to depend on $\varepsilon$. This order is the whole content of the definition, and reversing it produces a statement that is both different and false.

Here is the pattern of an $\varepsilon$-$N$ proof. Start from the inequality you must establish, solve it for $n$ on scratch paper, then write the proof forward, beginning with the $N$ that your scratch work produced.

__Claim.__ $\displaystyle \lim_{n \to \infty} \frac{1}{n} = 0$.

__Proof.__ Let $\varepsilon > 0$ be given. By the Archimedean property of $\mathbb{R}$, which follows from completeness, there is a positive integer $N$ with $N > 1/\varepsilon$. Suppose $n > N$. Then $n > 1/\varepsilon$, and since $n$ and $\varepsilon$ are positive this gives $1/n < \varepsilon$. Therefore

$$
\left| \frac{1}{n} - 0 \right| = \frac{1}{n} < \varepsilon
$$

Since $\varepsilon > 0$ was arbitrary, the sequence converges to $0$. $\blacksquare$

Notice what the proof does not do. It never chooses a particular $\varepsilon$, it never says the terms get close to $0$, and it never appeals to a picture. It exhibits an $N$ for an arbitrary $\varepsilon$, which is exactly what the definition demands.

#### <a style="text-decoration: none" class="anchor" id="Practice:P4EpsilonN">Practice Exercise 1 - An Epsilon-N Proof</a>

Prove directly from the definition that

$$
\lim_{n \to \infty} \frac{n+1}{n} = 1
$$

<details>
    <br>
    <summary>
        View Solution
    </summary>

__Scratch work.__ We need $\left| \dfrac{n+1}{n} - 1 \right| < \varepsilon$. Simplify the quantity first:

$$
\left| \frac{n+1}{n} - 1 \right| = \left| \frac{n+1-n}{n} \right| = \frac{1}{n}
$$

so the requirement is $1/n < \varepsilon$, which holds exactly when $n > 1/\varepsilon$. That tells us which $N$ to choose. Now write the proof forward.

__Proof.__ Let $\varepsilon > 0$ be given. Choose a positive integer $N$ with $N > 1/\varepsilon$, which is possible by the Archimedean property. Let $n > N$. Then $n > 1/\varepsilon$, and since $n > 0$ and $\varepsilon > 0$ we may multiply both sides by $\varepsilon/n$ to obtain $\varepsilon > 1/n$. Hence

$$
\left| \frac{n+1}{n} - 1 \right| = \frac{1}{n} < \varepsilon
$$

Since $\varepsilon > 0$ was arbitrary, $\lim_{n\to\infty} (n+1)/n = 1$. $\blacksquare$

The scratch work is not part of the proof and should not appear in your submission, but you should always do it. Every $\varepsilon$-$N$ proof is written backwards and read forwards.
</details>

### Quantifiers Come First

Before proving a claim you must know its logical shape, because the shape dictates what a proof and what a disproof look like.

<div class="alert alert-block alert-info">

#### <a style="text-decoration: none" class="anchor" id="Def:P4Quantifiers">Definition</a>

__Quantifiers__

A __universal__ statement asserts that a property holds for every object of some kind, written $\forall$. An __existential__ statement asserts that at least one object has the property, written $\exists$.

$$
\begin{array}{|l|l|l|}
    \hline
    \textbf{Statement} & \textbf{To prove it} & \textbf{To disprove it} \\
    \hline
    \forall x,\ P(x) & \text{argue for an arbitrary } x & \text{exhibit one } x \text{ with } P(x) \text{ false} \\
    \hline
    \exists x,\ P(x) & \text{exhibit one such } x & \text{argue no } x \text{ can work} \\
    \hline
\end{array}
$$

The negation of $\forall x,\ P(x)$ is $\exists x,\ \neg P(x)$, and the negation of $\exists x,\ P(x)$ is $\forall x,\ \neg P(x)$.
</div>

Mathematical writing hides quantifiers constantly. "A matrix with more columns than rows has a nonzero solution to $A\mathbf{x} = \mathbf{0}$" means: for every field $F$, for every $m$ and $n$ with $n > m$, and for every $m \times n$ matrix $A$ over $F$, there exists a nonzero $\mathbf{x}$ with $A\mathbf{x} = \mathbf{0}$. Two universals and an existential. Your proof must begin by taking an arbitrary such $A$ and must end by producing a specific $\mathbf{x}$.

<div class="alert alert-block alert-info">

#### <a style="text-decoration: none" class="anchor" id="Def:P4Forms">Definition</a>

__Four Forms of Proof__

__Direct proof.__ To prove "if $P$ then $Q$", assume $P$ and derive $Q$ by definitions, axioms, and established results.

__Counterexample.__ To disprove a universal statement, exhibit a single object for which it fails. One counterexample settles the matter permanently.

__Double inclusion.__ To prove two sets are equal, prove $S \subseteq T$ and $T \subseteq S$. To prove $S \subseteq T$, take an arbitrary element of $S$ and show it lies in $T$.

__Proof by cases.__ When the hypothesis splits into finitely many exhaustive situations, prove the conclusion in each. The cases must cover every possibility.
</div>

### Evidence Is Not Proof

In Practice Exercise 4 of Problem Set 1.P3 you built random integer matrices with `randi([0 9],6,6)` and observed their behavior. That habit is valuable and it is not proof. Consider the claim

$$
\text{every } 6 \times 6 \text{ matrix whose entries are integers from } 0 \text{ to } 9 \text{ is nonsingular}
$$

Generate ten thousand such matrices in MATLAB and every one of them will very likely be nonsingular. The evidence is overwhelming and the claim is false: the matrix of all zeros has integer entries from $0$ to $9$ and is singular, as does any such matrix with two equal rows. The claim is universal, so a single counterexample destroys it, and no quantity of random trials can establish it.

The reverse error is just as common. A claim that fails on a hundred random trials is false, and that observation __is__ a disproof, provided you exhibit one failing case explicitly rather than reporting a failure rate.

#### <a style="text-decoration: none" class="anchor" id="Practice:P4Evidence">Practice Exercise 2 - Evidence Is Not Proof</a>

Consider the claim: for every $2 \times 2$ matrix $A$ all of whose entries are positive real numbers, the system $A\mathbf{x} = \mathbf{b}$ has a unique solution for every $\mathbf{b} \in \mathbb{R}^2$.

(a) Write the claim with its quantifiers made explicit.

(b) Decide whether the claim is true. Prove it or disprove it.

<details>
    <br>
    <summary>
        View Solution
    </summary>

(a) For every $A \in \mathbb{R}^{2\times 2}$ such that $a_{ij} > 0$ for all $i$ and $j$, and for every $\mathbf{b} \in \mathbb{R}^2$, there exists exactly one $\mathbf{x} \in \mathbb{R}^2$ with $A\mathbf{x} = \mathbf{b}$. The claim is universal in both $A$ and $\mathbf{b}$, so a single choice of $A$ and $\mathbf{b}$ for which it fails disproves it.

(b) The claim is false. Take

$$
A = \begin{bmatrix} 1 & 1 \\ 1 & 1 \end{bmatrix}, \qquad \mathbf{b} = \begin{bmatrix} 1 \\ 0 \end{bmatrix}
$$

Every entry of $A$ is positive. If $A\mathbf{x} = \mathbf{b}$ then the first equation gives $x_1 + x_2 = 1$ and the second gives $x_1 + x_2 = 0$, so $1 = 0$, a contradiction. Hence no solution exists and the claim fails. $\blacksquare$

Note that a second counterexample is available with $\mathbf{b} = (1,1)^T$, where the system is consistent but has infinitely many solutions, so uniqueness fails as well. Either one suffices; there is no need to give both.

Observe also that positivity of the entries was a distraction. It is a hypothesis that sounds restrictive and controls nothing relevant, and recognizing that is part of reading a claim carefully.
</details>

### Definitions Made Precise

Section 1.1.10 introduced linear combinations and span through pictures. Here are the same ideas stated so that they can be used in a proof. Throughout, $F$ denotes $\mathbb{R}$ or $\mathbb{C}$.

<div class="alert alert-block alert-info">

#### <a style="text-decoration: none" class="anchor" id="Def:P4Span">Definition</a>

__Linear Combination and Span__

Let $\mathbf{v}_1, \mathbf{v}_2, \ldots, \mathbf{v}_k$ be vectors in $F^n$. A __linear combination__ of these vectors is any vector of the form

$$
c_1\mathbf{v}_1 + c_2\mathbf{v}_2 + \cdots + c_k\mathbf{v}_k, \qquad c_1, \ldots, c_k \in F
$$

The __span__ of the vectors is the set of all such vectors,

$$
\operatorname{span}\{\mathbf{v}_1, \ldots, \mathbf{v}_k\} = \left\{\, c_1\mathbf{v}_1 + \cdots + c_k\mathbf{v}_k \,:\, c_1, \ldots, c_k \in F \,\right\}
$$
</div>

The definition is a set built by a rule, so proving that a vector belongs to a span means exhibiting the scalars, and proving that a set is contained in a span means producing scalars for an arbitrary element of that set.

The first theorem of this problem set connects the definition to the computation you performed all through Chapter One. It is stated as a biconditional, so its proof has two halves, and each half is a direct proof.

<div class="alert alert-block alert-info">

#### <a style="text-decoration: none" class="anchor" id="Thm:P4Consistency">Theorem 1</a>

__Consistency and Span__

Let $A$ be an $m \times n$ matrix over $F$ with columns $\mathbf{a}_1, \ldots, \mathbf{a}_n$, and let $\mathbf{b} \in F^m$. Then the system $A\mathbf{x} = \mathbf{b}$ is consistent if and only if $\mathbf{b} \in \operatorname{span}\{\mathbf{a}_1, \ldots, \mathbf{a}_n\}$.
</div>

__Proof.__ Recall from Section 1.2.1 that for any $\mathbf{x} = (x_1, \ldots, x_n)^T$ in $F^n$,

$$
A\mathbf{x} = x_1\mathbf{a}_1 + x_2\mathbf{a}_2 + \cdots + x_n\mathbf{a}_n
$$

$(\Rightarrow)$ Suppose the system is consistent. Then there exists $\mathbf{x} \in F^n$ with $A\mathbf{x} = \mathbf{b}$. By the identity above, $\mathbf{b} = x_1\mathbf{a}_1 + \cdots + x_n\mathbf{a}_n$, which exhibits $\mathbf{b}$ as a linear combination of the columns with scalars $x_1, \ldots, x_n$. Hence $\mathbf{b}$ lies in the span.

$(\Leftarrow)$ Suppose $\mathbf{b} \in \operatorname{span}\{\mathbf{a}_1, \ldots, \mathbf{a}_n\}$. By definition there exist scalars $c_1, \ldots, c_n \in F$ with $\mathbf{b} = c_1\mathbf{a}_1 + \cdots + c_n\mathbf{a}_n$. Set $\mathbf{x} = (c_1, \ldots, c_n)^T$. Then $A\mathbf{x} = c_1\mathbf{a}_1 + \cdots + c_n\mathbf{a}_n = \mathbf{b}$, so the system has a solution and is consistent. $\blacksquare$

The two halves are mirror images: one direction converts a solution into scalars, the other converts scalars into a solution. That is not a coincidence but the content of the theorem, and it is why the column picture and the row-reduction procedure answer the same question.

#### <a style="text-decoration: none" class="anchor" id="Practice:P4Closure">Practice Exercise 3 - Closure of a Span</a>

Let $W = \operatorname{span}\{\mathbf{v}_1, \ldots, \mathbf{v}_k\}$ in $F^n$. Prove that if $\mathbf{u} \in W$ and $\mathbf{w} \in W$ and $c \in F$, then $\mathbf{u} + \mathbf{w} \in W$ and $c\mathbf{u} \in W$.

<details>
    <br>
    <summary>
        View Solution
    </summary>

__Proof.__ Let $\mathbf{u}, \mathbf{w} \in W$ and let $c \in F$ be arbitrary. By the definition of span there exist scalars $a_1, \ldots, a_k$ and $b_1, \ldots, b_k$ in $F$ with

$$
\mathbf{u} = a_1\mathbf{v}_1 + \cdots + a_k\mathbf{v}_k, \qquad \mathbf{w} = b_1\mathbf{v}_1 + \cdots + b_k\mathbf{v}_k
$$

Adding and collecting the coefficient of each $\mathbf{v}_i$,

$$
\mathbf{u} + \mathbf{w} = (a_1 + b_1)\mathbf{v}_1 + \cdots + (a_k + b_k)\mathbf{v}_k
$$

Each $a_i + b_i$ lies in $F$, because $F$ is closed under addition, so $\mathbf{u} + \mathbf{w}$ is a linear combination of $\mathbf{v}_1, \ldots, \mathbf{v}_k$ and therefore lies in $W$.

Similarly,

$$
c\mathbf{u} = c(a_1\mathbf{v}_1 + \cdots + a_k\mathbf{v}_k) = (ca_1)\mathbf{v}_1 + \cdots + (ca_k)\mathbf{v}_k
$$

where the second equality uses distributivity and the associativity of multiplication in $F$. Each $ca_i$ lies in $F$, so $c\mathbf{u} \in W$. Since $\mathbf{u}$, $\mathbf{w}$, and $c$ were arbitrary, the claim holds. $\blacksquare$

The proof used two facts about $F$ and nothing else: that sums and products of scalars are again scalars, and that multiplication distributes over addition. Everything else is bookkeeping.
</details>

### The Free Variables of Chapter One, Made Into Theorems

In Exercises 2, 6, 7, and 8 of Problem Set 1.P3 you computed reduced row echelon forms, read off free variables, and wrote solution sets in the form $\mathbf{w} + s\mathbf{z}_1 + t\mathbf{z}_2$. Every one of those computations relied on a structural fact that was demonstrated by example and never proved. Here it is proved.

Throughout, $A$ is an $m \times n$ matrix over $F$ and

$$
N(A) = \left\{\, \mathbf{z} \in F^n \,:\, A\mathbf{z} = \mathbf{0} \,\right\}
$$

denotes the set of solutions of the homogeneous system.

<div class="alert alert-block alert-info">

#### <a style="text-decoration: none" class="anchor" id="Thm:P4Homogeneous">Theorem 2</a>

__Homogeneous Solutions Are Closed__

If $\mathbf{z}_1, \mathbf{z}_2 \in N(A)$ and $c_1, c_2 \in F$, then $c_1\mathbf{z}_1 + c_2\mathbf{z}_2 \in N(A)$. In particular $\mathbf{0} \in N(A)$.
</div>

__Proof.__ Let $\mathbf{z}_1, \mathbf{z}_2 \in N(A)$ and $c_1, c_2 \in F$. Using the linearity of the matrix-vector product, proved as Theorem 5 in the next section,

$$
A(c_1\mathbf{z}_1 + c_2\mathbf{z}_2) = c_1(A\mathbf{z}_1) + c_2(A\mathbf{z}_2) = c_1\mathbf{0} + c_2\mathbf{0} = \mathbf{0}
$$

so $c_1\mathbf{z}_1 + c_2\mathbf{z}_2 \in N(A)$. Taking $c_1 = c_2 = 0$ shows $\mathbf{0} \in N(A)$. $\blacksquare$

<div class="alert alert-block alert-info">

#### <a style="text-decoration: none" class="anchor" id="Thm:P4Structure">Theorem 3</a>

__Structure of a Solution Set__

Suppose $A\mathbf{x} = \mathbf{b}$ is consistent and let $\mathbf{w}$ be any particular solution. Then the set of all solutions is

$$
S = \left\{\, \mathbf{w} + \mathbf{z} \,:\, \mathbf{z} \in N(A) \,\right\}
$$
</div>

__Proof.__ This is an equality of sets, so we prove two inclusions.

$(\supseteq)$ Let $\mathbf{z} \in N(A)$. Then

$$
A(\mathbf{w} + \mathbf{z}) = A\mathbf{w} + A\mathbf{z} = \mathbf{b} + \mathbf{0} = \mathbf{b}
$$

so $\mathbf{w} + \mathbf{z}$ is a solution.

$(\subseteq)$ Let $\mathbf{x}$ be any solution, so $A\mathbf{x} = \mathbf{b}$. Put $\mathbf{z} = \mathbf{x} - \mathbf{w}$. Then

$$
A\mathbf{z} = A(\mathbf{x} - \mathbf{w}) = A\mathbf{x} - A\mathbf{w} = \mathbf{b} - \mathbf{b} = \mathbf{0}
$$

so $\mathbf{z} \in N(A)$, and $\mathbf{x} = \mathbf{w} + \mathbf{z}$ has the required form.

Since each set contains the other, they are equal. $\blacksquare$

Notice that the theorem says nothing about which particular solution $\mathbf{w}$ you choose. Any solution works, and different choices produce the same set $S$ described differently. That is why two students who assign free variables differently can both be right.

<div class="alert alert-block alert-info">

#### <a style="text-decoration: none" class="anchor" id="Thm:P4Uniqueness">Theorem 4</a>

__Uniqueness__

For every $\mathbf{b} \in F^m$ the system $A\mathbf{x} = \mathbf{b}$ has at most one solution if and only if $N(A) = \{\mathbf{0}\}$.
</div>

The proof is Exercise 7. Together with Theorem 3, this is the precise version of the rule you have been using since Section 1.2.9: a free column produces a nonzero homogeneous solution, a nonzero homogeneous solution produces infinitely many solutions whenever any solution exists, and only a system with no free columns can have a unique solution.

### Proving What the Notation Suggests

Theorems 2 and 3 both used the rule $A(\mathbf{u} + \mathbf{v}) = A\mathbf{u} + A\mathbf{v}$. That rule is not an axiom and not a convention. It is a theorem about the definition of the matrix-vector product, and its proof is a computation with the $i$th entry in which every step is a field axiom.

<div class="alert alert-block alert-info">

#### <a style="text-decoration: none" class="anchor" id="Thm:P4Linearity">Theorem 5</a>

__Linearity of the Matrix-Vector Product__

Let $A$ be an $m \times n$ matrix over $F$, let $\mathbf{u}, \mathbf{v} \in F^n$, and let $c \in F$. Then

$$
A(\mathbf{u} + \mathbf{v}) = A\mathbf{u} + A\mathbf{v} \qquad \text{and} \qquad A(c\mathbf{u}) = c(A\mathbf{u})
$$
</div>

__Proof.__ Two vectors in $F^m$ are equal exactly when their corresponding entries are equal, so it suffices to fix $i$ with $1 \leq i \leq m$ and compare $i$th entries. By the definition of the matrix-vector product, the $i$th entry of $A\mathbf{x}$ is $\sum_{j=1}^{n} a_{ij}x_j$.

$$
\begin{align*}
    \left[A(\mathbf{u} + \mathbf{v})\right]_i
        &= \sum_{j=1}^{n} a_{ij}(u_j + v_j) && \text{definition of the product and of } \mathbf{u}+\mathbf{v} \\
        &= \sum_{j=1}^{n} \left( a_{ij}u_j + a_{ij}v_j \right) && \text{distributivity in } F \\
        &= \sum_{j=1}^{n} a_{ij}u_j + \sum_{j=1}^{n} a_{ij}v_j && \text{associativity and commutativity of addition} \\
        &= \left[A\mathbf{u}\right]_i + \left[A\mathbf{v}\right]_i && \text{definition of the product}
\end{align*}
$$

Since $i$ was arbitrary, the two vectors agree in every entry and are therefore equal. For the second identity,

$$
\begin{align*}
    \left[A(c\mathbf{u})\right]_i
        &= \sum_{j=1}^{n} a_{ij}(cu_j) && \text{definition of the product and of } c\mathbf{u} \\
        &= \sum_{j=1}^{n} c\,(a_{ij}u_j) && \text{associativity and } \textbf{commutativity} \text{ of multiplication} \\
        &= c \sum_{j=1}^{n} a_{ij}u_j && \text{distributivity} \\
        &= c\left[A\mathbf{u}\right]_i && \text{definition of the product}
\end{align*}
$$

Again $i$ was arbitrary, so the identity holds. $\blacksquare$

<div class="alert alert-block alert-success">

#### The Step That Needs Commutativity

Every step of the first identity uses only distributivity and the arithmetic of addition. The second identity is different. Moving $c$ from the right of $a_{ij}$ to its left, in the line marked above, requires $a_{ij}c = c\,a_{ij}$, and that is the commutative law of multiplication.

For scalars in $\mathbb{R}$ or $\mathbb{C}$ this is automatic and easy to overlook. There is a fourth number system, the quaternions, in which multiplication is associative and every nonzero element has an inverse but multiplication is __not__ commutative: $ij = k$ while $ji = -k$. Quaternions are how rotations are represented in robotics, aerospace attitude control, and computer graphics, so the question is not idle. Over such a system the identity $A(c\mathbf{u}) = c(A\mathbf{u})$ genuinely fails, and one must decide once and for all whether scalars multiply vectors from the left or from the right. We return to this at the end of the semester.
</div>


#### <a style="text-decoration: none" class="anchor" id="Practice:P4NameAxiom">Practice Exercise 5 - Name the Axiom</a>

Let $A$ be $m \times n$ over $F$, let $\mathbf{u} \in F^n$, and let $c, d \in F$. Prove that $A\big((c+d)\mathbf{u}\big) = c(A\mathbf{u}) + d(A\mathbf{u})$, naming the field axiom used at each step.

<details>
    <br>
    <summary>
        View Solution
    </summary>

__Proof.__ Fix $i$ with $1 \leq i \leq m$ and compare $i$th entries.

$$
\begin{align*}
    \left[A\big((c+d)\mathbf{u}\big)\right]_i
        &= \sum_{j=1}^{n} a_{ij}\big((c+d)u_j\big) && \text{definition of the product} \\
        &= \sum_{j=1}^{n} a_{ij}\big(cu_j + du_j\big) && \text{distributivity in } F \\
        &= \sum_{j=1}^{n} \left( a_{ij}(cu_j) + a_{ij}(du_j) \right) && \text{distributivity in } F \\
        &= \sum_{j=1}^{n} \left( c(a_{ij}u_j) + d(a_{ij}u_j) \right) && \text{associativity and commutativity of multiplication} \\
        &= c\sum_{j=1}^{n} a_{ij}u_j + d\sum_{j=1}^{n} a_{ij}u_j && \text{distributivity, associativity and commutativity of addition} \\
        &= c\left[A\mathbf{u}\right]_i + d\left[A\mathbf{u}\right]_i && \text{definition of the product}
\end{align*}
$$

Since $i$ was arbitrary the vectors are equal in every entry, hence equal. $\blacksquare$

An acceptable shorter proof cites Theorem 5 twice: $(c+d)\mathbf{u} = c\mathbf{u} + d\mathbf{u}$ entrywise, then $A(c\mathbf{u} + d\mathbf{u}) = A(c\mathbf{u}) + A(d\mathbf{u}) = c(A\mathbf{u}) + d(A\mathbf{u})$. Building on established results rather than returning to the axioms every time is not laziness; it is how mathematics is actually written. Return to the axioms when you are proving the foundational result, and cite it thereafter.
</details>

### Why the Completeness Axiom Was Stated

Nothing proved in 1.P4.3 through 1.P4.5 used completeness. Elimination, spans, and solution sets are finite algebra, and they work over any field. Completeness enters the moment a question involves infinitely many steps, and such questions arrive almost immediately.

<div class="alert alert-block alert-info">

#### <a style="text-decoration: none" class="anchor" id="Def:P4VectorLimits">Definition</a>

__Convergence of Sequences of Vectors and Matrices__

A sequence $(\mathbf{x}^{(k)})$ of vectors in $\mathbb{R}^n$ __converges entrywise__ to $\mathbf{x} \in \mathbb{R}^n$ if

$$
\lim_{k \to \infty} x^{(k)}_i = x_i \qquad \text{for each } i = 1, 2, \ldots, n
$$

A sequence $(A^{(k)})$ of $m \times n$ matrices __converges entrywise__ to $A$ if $\lim_{k\to\infty} a^{(k)}_{ij} = a_{ij}$ for every $i$ and $j$. Each of these is finitely many ordinary limits of real sequences, so each rests on the completeness axiom of 1.P4.1.
</div>

This definition is stated here and used in Chapter Two. Three places where it will matter:

__Markov matrices.__ Chapter Two studies transition matrices and their powers $P^k$. The question "what is the long-run distribution" is the question of whether the sequence $(P^k)$ converges entrywise, and if so to what. The answer is a theorem, not an observation about a table of numbers, and it cannot even be stated without the definition above.

__Little $\ell^2$.__ Convergence in $\mathbb{R}^n$ can be phrased through the quantity $\|\mathbf{x}\| = \sqrt{x_1^2 + \cdots + x_n^2}$. Allowing infinitely many entries with $\sum_{i=1}^{\infty} x_i^2$ finite gives the space $\ell^2$, where completeness is no longer a convenience but the property that makes the theory work at all.

__Fourier convergence.__ Writing a signal as an infinite sum of sinusoids asks whether a sequence of finite approximations converges, and in what sense. You met the finite version of this question in Exercise 3 of Problem Set 1.P3, where a sampled signal was decomposed into finitely many known frequencies by solving a linear system. The infinite version is the same idea with a limit attached, and every result about it depends on the completeness of the underlying space.

<div class="alert alert-block alert-success">

#### The Line Between the Two Halves of the Course

Everything in Chapter One is finite: finitely many equations, finitely many unknowns, finitely many row operations. Such results hold over any field and require no analysis whatsoever.

The moment the course asks what happens in the long run, in the limit, or for an infinite sum, the algebra is no longer sufficient and the completeness of $\mathbb{R}$ is doing the work. Knowing which side of that line a theorem lives on is a large part of understanding the subject.
</div>

__Topic: Axioms of $\mathbb{R}$.__

Prove each statement using only the field axioms of 1.P4.1. Do not use any property of the real numbers that you have not proved or been given as an axiom, and name the axiom at each step.

(a) The additive identity is unique: if $0'$ satisfies $a + 0' = a$ for every $a \in \mathbb{R}$, then $0' = 0$.

<details>
    <br>
    <summary>
        View Solution
    </summary>

__Proof.__ Suppose $0'$ satisfies $a + 0' = a$ for every real $a$. Taking $a = 0$ gives $0 + 0' = 0$. On the other hand, the additive identity axiom applied to $0'$ gives $0' + 0 = 0'$, and addition is commutative, so $0 + 0' = 0'$. Therefore

$$
0' = 0 + 0' = 0
$$

as required. $\blacksquare$

The structure is worth noting: we used the defining property of $0'$ on the element $0$, and the defining property of $0$ on the element $0'$, then compared. Every uniqueness proof of this kind has that shape.
</details>
<br>


(b) For every $a \in \mathbb{R}$, $\ 0 \cdot a = 0$.

<details>
    <br>
    <summary>
        View Solution
    </summary>

__Proof.__ Let $a \in \mathbb{R}$. Since $0 + 0 = 0$ by the additive identity axiom,

$$
0 \cdot a = (0 + 0)a = 0 \cdot a + 0 \cdot a
$$

using distributivity together with commutativity of multiplication. The number $0 \cdot a$ has an additive inverse $-(0\cdot a)$. Adding it to both sides,

$$
0 \cdot a + \big({-}(0 \cdot a)\big) = \big(0 \cdot a + 0 \cdot a\big) + \big({-}(0 \cdot a)\big)
$$

The left side is $0$. Using associativity of addition on the right side gives $0 \cdot a + \big(0\cdot a + ({-}(0\cdot a))\big) = 0 \cdot a + 0 = 0 \cdot a$. Hence $0 = 0 \cdot a$. $\blacksquare$

This statement looks too obvious to prove, which is exactly why it must be proved: it is not one of the axioms, and it does not follow from any single one of them.
</details>
<br>


(c) For all $a, b \in \mathbb{R}$, if $ab = 0$ then $a = 0$ or $b = 0$. Identify precisely which axiom fails if this is attempted in a system without multiplicative inverses.

<details>
    <br>
    <summary>
        View Solution
    </summary>

__Proof.__ Suppose $ab = 0$. If $a = 0$ the conclusion holds, so assume $a \neq 0$. Then $a^{-1}$ exists by the multiplicative inverse axiom, and

$$
b = 1 \cdot b = (a^{-1}a)b = a^{-1}(ab) = a^{-1} \cdot 0 = 0
$$

using the multiplicative identity, the inverse axiom, associativity of multiplication, the hypothesis, and part (b) in that order. Hence $b = 0$. In either case $a = 0$ or $b = 0$. $\blacksquare$

The proof is by cases on whether $a = 0$, and the second case is the only one that does any work.

The axiom that carries the argument is the existence of the multiplicative inverse $a^{-1}$. In a system where nonzero elements need not have inverses the conclusion can fail. In the integers modulo $6$, for instance, $2 \cdot 3 = 0$ although neither factor is $0$; neither $2$ nor $3$ has a multiplicative inverse there.
</details>
<br>


(d) For every $a \in \mathbb{R}$, $\ (-1)a = -a$, where $-a$ denotes the additive inverse of $a$.

<details>
    <br>
    <summary>
        View Solution
    </summary>

__Proof.__ Let $a \in \mathbb{R}$. It suffices to show that $(-1)a$ is an additive inverse of $a$, since additive inverses are unique. Compute

$$
a + (-1)a = 1 \cdot a + (-1)a = \big(1 + (-1)\big)a = 0 \cdot a = 0
$$

using the multiplicative identity axiom, distributivity with commutativity of multiplication, the additive inverse axiom, and part (b). Hence $(-1)a$ is an additive inverse of $a$, so $(-1)a = -a$. $\blacksquare$

Strictly, this argument uses the uniqueness of additive inverses, which is proved exactly as in part (a): if $b$ and $b'$ are both additive inverses of $a$ then $b = b + 0 = b + (a + b') = (b+a) + b' = 0 + b' = b'$.
</details>

__Topic: Completeness.__

Let $S = \{\, x \in \mathbb{Q} : x > 0 \text{ and } x^2 < 2 \,\}$. This exercise proves that $S$ has no least upper bound in $\mathbb{Q}$, so the rational numbers fail the completeness axiom. Throughout, you may use without proof that there is no rational number $p$ with $p^2 = 2$.

(a) Show that $S$ is nonempty and bounded above in $\mathbb{Q}$.

<details>
    <br>
    <summary>
        View Solution
    </summary>

__Proof.__ The number $1$ is rational, positive, and satisfies $1^2 = 1 < 2$, so $1 \in S$ and $S$ is nonempty.

The number $2$ is an upper bound. Suppose $x \in S$ with $x > 2$. Then, since $x > 2 > 0$, multiplying the inequality $x > 2$ by the positive number $x$ gives $x^2 > 2x > 4 > 2$, contradicting $x^2 < 2$. Hence $x \leq 2$ for every $x \in S$. $\blacksquare$
</details>
<br>


(b) For a positive rational $p$, define

$$
q = p - \frac{p^2 - 2}{p + 2} = \frac{2p + 2}{p + 2}
$$

Verify that $q$ is rational and positive, and show that

$$
q^2 - 2 = \frac{2(p^2 - 2)}{(p+2)^2}
$$

<details>
    <br>
    <summary>
        View Solution
    </summary>

__Proof.__ Since $p$ is rational and positive, $p + 2$ is a nonzero rational, so $q = (2p+2)/(p+2)$ is a quotient of rationals with nonzero denominator and is therefore rational. Both numerator and denominator are positive, so $q > 0$.

The two expressions for $q$ agree:

$$
p - \frac{p^2-2}{p+2} = \frac{p(p+2) - (p^2-2)}{p+2} = \frac{p^2 + 2p - p^2 + 2}{p+2} = \frac{2p+2}{p+2}
$$

Now compute

$$
q^2 - 2 = \frac{(2p+2)^2}{(p+2)^2} - 2 = \frac{(2p+2)^2 - 2(p+2)^2}{(p+2)^2}
$$

The numerator is $\left(4p^2 + 8p + 4\right) - \left(2p^2 + 8p + 8\right) = 2p^2 - 4 = 2(p^2 - 2)$, which gives the stated identity. $\blacksquare$
</details>
<br>


(c) Use part (b) to show that if $p \in S$ then $q \in S$ and $q > p$; and that if $p$ is a positive rational upper bound for $S$ with $p^2 > 2$, then $q$ is also an upper bound with $q < p$.

<details>
    <br>
    <summary>
        View Solution
    </summary>

__Proof.__ Suppose $p \in S$, so $p > 0$ and $p^2 < 2$. Then $p^2 - 2 < 0$, so by part (b) $q^2 - 2 < 0$, giving $q^2 < 2$, and $q > 0$ was shown in (b). Hence $q \in S$. Moreover $q - p = -(p^2-2)/(p+2) > 0$ because the numerator is positive and the denominator is positive, so $q > p$.

Now suppose $p$ is a positive rational with $p^2 > 2$. Then $p^2 - 2 > 0$, so $q^2 - 2 > 0$ by part (b), giving $q^2 > 2$, and $q - p = -(p^2-2)/(p+2) < 0$, so $q < p$.

It remains to see that any positive rational $r$ with $r^2 > 2$ is an upper bound for $S$. If $x \in S$ then $x^2 < 2 < r^2$, so $r^2 - x^2 = (r-x)(r+x) > 0$; since $r + x > 0$ this forces $r - x > 0$, that is, $x < r$. Applying this to $q$, which satisfies $q > 0$ and $q^2 > 2$, shows $q$ is an upper bound for $S$. $\blacksquare$
</details>
<br>


(d) Conclude that $S$ has no least upper bound in $\mathbb{Q}$. Explain in a sentence why this does not contradict the completeness axiom.

<details>
    <br>
    <summary>
        View Solution
    </summary>

__Proof.__ Suppose, for contradiction, that $p \in \mathbb{Q}$ is a least upper bound for $S$. Since $1 \in S$ we have $p \geq 1 > 0$, so $p$ is a positive rational. By trichotomy exactly one of $p^2 < 2$, $p^2 = 2$, $p^2 > 2$ holds.

If $p^2 = 2$, then $p$ is a rational whose square is $2$, which does not exist.

If $p^2 < 2$, then $p \in S$, and part (c) produces $q \in S$ with $q > p$. Then $p$ is not an upper bound for $S$, a contradiction.

If $p^2 > 2$, then part (c) produces an upper bound $q$ for $S$ with $q < p$. Then $p$ is not a __least__ upper bound, a contradiction.

All three cases are impossible, so no such $p$ exists. $\blacksquare$

There is no contradiction with the completeness axiom, because that axiom is an assertion about subsets of $\mathbb{R}$, not of $\mathbb{Q}$. The set $S$ does have a least upper bound in $\mathbb{R}$, namely $\sqrt{2}$; what this exercise shows is that the bound is not rational, so $\mathbb{Q}$ has a gap exactly where the completeness axiom would require a number to be.
</details>

__Topic: Epsilon-N Proofs.__

Prove each limit directly from the definition in 1.P4.1. Show your scratch work separately from the proof, and make sure the proof itself reads forward.

(a) $\displaystyle \lim_{n \to \infty} \frac{3n+1}{n+2} = 3$

<details>
    <br>
    <summary>
        View Solution
    </summary>

__Scratch work.__

$$
\left| \frac{3n+1}{n+2} - 3 \right| = \left| \frac{3n + 1 - 3n - 6}{n+2} \right| = \frac{5}{n+2}
$$

and $\dfrac{5}{n+2} < \varepsilon$ exactly when $n + 2 > 5/\varepsilon$, that is, $n > 5/\varepsilon - 2$.

__Proof.__ Let $\varepsilon > 0$ be given. Choose a positive integer $N$ with $N > 5/\varepsilon$, possible by the Archimedean property. Let $n > N$. Then $n + 2 > n > 5/\varepsilon > 0$, so $\dfrac{5}{n+2} < \varepsilon$. Therefore

$$
\left| \frac{3n+1}{n+2} - 3 \right| = \frac{5}{n+2} < \varepsilon
$$

Since $\varepsilon > 0$ was arbitrary, the limit is $3$. $\blacksquare$

Choosing $N > 5/\varepsilon$ rather than $N > 5/\varepsilon - 2$ is legitimate and simpler. Any $N$ that works is acceptable; there is no requirement that it be the smallest one.
</details>
<br>


(b) $\displaystyle \lim_{n \to \infty} \frac{1}{2^n} = 0$. You may use without proof that $2^n > n$ for every positive integer $n$.

<details>
    <br>
    <summary>
        View Solution
    </summary>

__Proof.__ Let $\varepsilon > 0$ be given. Choose a positive integer $N$ with $N > 1/\varepsilon$. Let $n > N$. Then $2^n > n > N > 1/\varepsilon > 0$, and taking reciprocals of the positive numbers $2^n$ and $1/\varepsilon$ reverses the inequality, giving $\dfrac{1}{2^n} < \varepsilon$. Hence

$$
\left| \frac{1}{2^n} - 0 \right| = \frac{1}{2^n} < \varepsilon
$$

Since $\varepsilon > 0$ was arbitrary, the limit is $0$. $\blacksquare$

This sequence is the simplest example of the behavior Chapter Two will study in Markov matrices, where entries of $P^k$ approach their limits at a geometric rate of exactly this kind.
</details>
<br>


(c) Explain why the following is __not__ a proof that $\lim_{n\to\infty} 1/n = 0$: "Take $\varepsilon = 0.001$. Then for $n > 1000$ we have $1/n < 0.001$, so the terms get arbitrarily close to $0$."

<details>
    <br>
    <summary>
        View Solution
    </summary>

The argument fixes a particular $\varepsilon$. The definition requires that a suitable $N$ exist for __every__ positive $\varepsilon$, so verifying one value of $\varepsilon$, or any finite list of values, establishes nothing about the others. The quantifier structure is $\forall \varepsilon\, \exists N\, \forall n > N$, and the given argument addresses only a single instance of the outermost quantifier.

The phrase "get arbitrarily close" is also doing unearned work. It is an informal restatement of the conclusion rather than a step in an argument, and a proof may not appeal to it.

The repair is to let $\varepsilon > 0$ be arbitrary and produce $N$ as a formula in $\varepsilon$, as in the worked example of 1.P4.1.
</details>

__Topic: Span, Biconditional Proof.__

This exercise asks you to reproduce and then extend Theorem 1. Let $A$ be $m \times n$ over $F$ with columns $\mathbf{a}_1, \ldots, \mathbf{a}_n$.

(a) State Theorem 1 with all quantifiers explicit, and explain why proving a biconditional requires two separate arguments.

<details>
    <br>
    <summary>
        View Solution
    </summary>

For every $m \times n$ matrix $A$ over $F$ and every $\mathbf{b} \in F^m$: there exists $\mathbf{x} \in F^n$ with $A\mathbf{x} = \mathbf{b}$ if and only if there exist scalars $c_1, \ldots, c_n \in F$ with $\mathbf{b} = c_1\mathbf{a}_1 + \cdots + c_n\mathbf{a}_n$.

A biconditional $P \leftrightarrow Q$ asserts two implications, $P \to Q$ and $Q \to P$. Neither follows from the other; a statement can be true in one direction and false in the reverse. Each implication requires its own direct proof, beginning from its own hypothesis.
</details>
<br>


(b) Prove the direction: if $A\mathbf{x} = \mathbf{b}$ is consistent then $\mathbf{b} \in \operatorname{span}\{\mathbf{a}_1, \ldots, \mathbf{a}_n\}$.

<details>
    <br>
    <summary>
        View Solution
    </summary>

__Proof.__ Assume the system is consistent, so there exists $\mathbf{x} = (x_1, \ldots, x_n)^T \in F^n$ with $A\mathbf{x} = \mathbf{b}$. By the column form of the matrix-vector product,

$$
\mathbf{b} = A\mathbf{x} = x_1\mathbf{a}_1 + x_2\mathbf{a}_2 + \cdots + x_n\mathbf{a}_n
$$

Taking $c_i = x_i$ for each $i$ exhibits $\mathbf{b}$ as a linear combination of the columns with scalars in $F$, so $\mathbf{b}$ belongs to their span. $\blacksquare$
</details>
<br>


(c) Prove the converse.

<details>
    <br>
    <summary>
        View Solution
    </summary>

__Proof.__ Assume $\mathbf{b} \in \operatorname{span}\{\mathbf{a}_1, \ldots, \mathbf{a}_n\}$. By the definition of span there are scalars $c_1, \ldots, c_n \in F$ with

$$
\mathbf{b} = c_1\mathbf{a}_1 + \cdots + c_n\mathbf{a}_n
$$

Define $\mathbf{x} = (c_1, \ldots, c_n)^T \in F^n$. By the column form of the product, $A\mathbf{x} = c_1\mathbf{a}_1 + \cdots + c_n\mathbf{a}_n = \mathbf{b}$. Hence the system has a solution and is consistent. $\blacksquare$
</details>
<br>


(d) Use Theorem 1 to prove: if the columns of $A$ span all of $F^m$, then $A\mathbf{x} = \mathbf{b}$ is consistent for every $\mathbf{b} \in F^m$. Then state, without proof, what this says about the reduced row echelon form of $A$.

<details>
    <br>
    <summary>
        View Solution
    </summary>

__Proof.__ Assume $\operatorname{span}\{\mathbf{a}_1, \ldots, \mathbf{a}_n\} = F^m$. Let $\mathbf{b} \in F^m$ be arbitrary. Then $\mathbf{b}$ lies in the span of the columns, so by the direction of Theorem 1 proved in part (c), the system $A\mathbf{x} = \mathbf{b}$ is consistent. Since $\mathbf{b}$ was arbitrary, the system is consistent for every right-hand side. $\blacksquare$

In terms of echelon forms, the condition says that the reduced row echelon form of $A$ has a pivot in every row. If some row of the reduced form were entirely zero, one could choose $\mathbf{b}$ making the corresponding entry of the reduced augmented column nonzero, producing an inconsistent system.
</details>

</H2>

__Topic: Span, Double Inclusion.__

Let $\mathbf{v}_1, \ldots, \mathbf{v}_k \in F^n$ and write $W = \operatorname{span}\{\mathbf{v}_1, \ldots, \mathbf{v}_k\}$. Call a subset $U \subseteq F^n$ __closed__ if $\mathbf{u} + \mathbf{u}' \in U$ and $c\mathbf{u} \in U$ whenever $\mathbf{u}, \mathbf{u}' \in U$ and $c \in F$.

(a) Prove that $W$ is closed.

<details>
    <br>
    <summary>
        View Solution
    </summary>

__Proof.__ Let $\mathbf{u}, \mathbf{u}' \in W$ and $c \in F$. By definition of span there are scalars with

$$
\mathbf{u} = \sum_{i=1}^{k} a_i \mathbf{v}_i, \qquad \mathbf{u}' = \sum_{i=1}^{k} b_i \mathbf{v}_i
$$

Then $\mathbf{u} + \mathbf{u}' = \sum_{i=1}^{k} (a_i + b_i)\mathbf{v}_i$ and $c\mathbf{u} = \sum_{i=1}^{k} (ca_i)\mathbf{v}_i$, where each $a_i + b_i$ and each $ca_i$ lies in $F$ because $F$ is closed under its own operations. Both are therefore linear combinations of $\mathbf{v}_1, \ldots, \mathbf{v}_k$ and lie in $W$. $\blacksquare$
</details>
<br>


(b) Prove that $\mathbf{v}_j \in W$ for each $j$, and that $\mathbf{0} \in W$.

<details>
    <br>
    <summary>
        View Solution
    </summary>

__Proof.__ Fix $j$. Take $c_j = 1$ and $c_i = 0$ for $i \neq j$. Then

$$
c_1\mathbf{v}_1 + \cdots + c_k\mathbf{v}_k = \mathbf{v}_j
$$

using $1\mathbf{v}_j = \mathbf{v}_j$ and $0\mathbf{v}_i = \mathbf{0}$ entrywise, so $\mathbf{v}_j \in W$. Taking every $c_i = 0$ gives $\mathbf{0} \in W$. $\blacksquare$
</details>
<br>


(c) Prove that if $U$ is closed and $\mathbf{v}_i \in U$ for every $i$, then $W \subseteq U$. Conclude that $W$ is the smallest closed set containing all the $\mathbf{v}_i$.

<details>
    <br>
    <summary>
        View Solution
    </summary>

__Proof.__ Assume $U$ is closed and contains every $\mathbf{v}_i$. Let $\mathbf{u} \in W$ be arbitrary, so $\mathbf{u} = c_1\mathbf{v}_1 + \cdots + c_k\mathbf{v}_k$ for some scalars $c_i \in F$.

Each $c_i\mathbf{v}_i$ lies in $U$, because $\mathbf{v}_i \in U$ and $U$ is closed under scalar multiplication. Adding two elements of $U$ stays in $U$, so $c_1\mathbf{v}_1 + c_2\mathbf{v}_2 \in U$; adding $c_3\mathbf{v}_3$ keeps the sum in $U$, and continuing in this way through all $k$ terms, a finite number of steps, gives $\mathbf{u} \in U$. Since $\mathbf{u}$ was arbitrary, $W \subseteq U$. $\blacksquare$

Combining with parts (a) and (b): $W$ is a closed set containing every $\mathbf{v}_i$, and it is contained in every such set. That is precisely what it means to call $W$ the smallest closed set containing the vectors.

The phrase "continuing in this way" is an informal induction on the number of terms. Stating it as such is the fully rigorous version, and you should expect to write the induction explicitly in later courses.
</details>
<br>


(d) Show by counterexample that the union of two closed sets need not be closed. Take $F = \mathbb{R}$ and $n = 2$.

<details>
    <br>
    <summary>
        View Solution
    </summary>

Let $U = \operatorname{span}\{(1,0)^T\}$ and $V = \operatorname{span}\{(0,1)^T\}$, each closed by part (a). Consider $\mathbf{u} = (1,0)^T \in U$ and $\mathbf{v} = (0,1)^T \in V$, both elements of $U \cup V$. Their sum is

$$
\mathbf{u} + \mathbf{v} = \begin{bmatrix} 1 \\ 1 \end{bmatrix}
$$

If $(1,1)^T \in U$ then $(1,1)^T = c(1,0)^T$ for some real $c$, forcing the second entry $1 = 0$, which is false. Similarly $(1,1)^T \notin V$. Hence $(1,1)^T \notin U \cup V$, so the union is not closed under addition. $\blacksquare$

This is a genuine counterexample and not an accident of the choice: the union of two distinct lines through the origin is never closed.
</details>

__Topic: Solution Sets.__

Let $A$ be $m \times n$ over $F$ and $\mathbf{b} \in F^m$, and let $N(A) = \{\mathbf{z} \in F^n : A\mathbf{z} = \mathbf{0}\}$. You may use Theorem 5.

(a) Prove Theorem 2: $N(A)$ is closed under linear combinations, and $\mathbf{0} \in N(A)$.

<details>
    <br>
    <summary>
        View Solution
    </summary>

__Proof.__ Let $\mathbf{z}_1, \mathbf{z}_2 \in N(A)$ and $c_1, c_2 \in F$. By Theorem 5, applied twice for the scalar multiples and once for the sum,

$$
A(c_1\mathbf{z}_1 + c_2\mathbf{z}_2) = A(c_1\mathbf{z}_1) + A(c_2\mathbf{z}_2) = c_1(A\mathbf{z}_1) + c_2(A\mathbf{z}_2) = c_1\mathbf{0} + c_2\mathbf{0} = \mathbf{0}
$$

so $c_1\mathbf{z}_1 + c_2\mathbf{z}_2 \in N(A)$. Setting $c_1 = c_2 = 0$ gives $\mathbf{0} \in N(A)$; alternatively, $A\mathbf{0} = \mathbf{0}$ directly from the definition of the product. $\blacksquare$
</details>
<br>


(b) Prove Theorem 3 by double inclusion: if $\mathbf{w}$ is any particular solution of $A\mathbf{x} = \mathbf{b}$, then the solution set equals $\{\mathbf{w} + \mathbf{z} : \mathbf{z} \in N(A)\}$.

<details>
    <br>
    <summary>
        View Solution
    </summary>

__Proof.__ Write $S$ for the set of all solutions and $T = \{\mathbf{w} + \mathbf{z} : \mathbf{z} \in N(A)\}$.

$T \subseteq S$: let $\mathbf{z} \in N(A)$. By Theorem 5, $A(\mathbf{w} + \mathbf{z}) = A\mathbf{w} + A\mathbf{z} = \mathbf{b} + \mathbf{0} = \mathbf{b}$, so $\mathbf{w} + \mathbf{z} \in S$.

$S \subseteq T$: let $\mathbf{x} \in S$, so $A\mathbf{x} = \mathbf{b}$. Set $\mathbf{z} = \mathbf{x} - \mathbf{w}$. Then $A\mathbf{z} = A\mathbf{x} - A\mathbf{w} = \mathbf{b} - \mathbf{b} = \mathbf{0}$, so $\mathbf{z} \in N(A)$, and $\mathbf{x} = \mathbf{w} + \mathbf{z} \in T$.

Both inclusions hold, so $S = T$. $\blacksquare$
</details>
<br>


(c) Prove that the solution set of a consistent system $A\mathbf{x} = \mathbf{b}$ is closed under addition if and only if $\mathbf{b} = \mathbf{0}$.

<details>
    <br>
    <summary>
        View Solution
    </summary>

__Proof.__ $(\Leftarrow)$ If $\mathbf{b} = \mathbf{0}$ the solution set is $N(A)$, which is closed under addition by part (a).

$(\Rightarrow)$ Suppose the solution set $S$ is closed under addition, and let $\mathbf{w} \in S$, which exists because the system is consistent. Then $\mathbf{w} + \mathbf{w} \in S$, so

$$
\mathbf{b} = A(\mathbf{w} + \mathbf{w}) = A\mathbf{w} + A\mathbf{w} = \mathbf{b} + \mathbf{b}
$$

Adding the additive inverse of $\mathbf{b}$ to both sides gives $\mathbf{0} = \mathbf{b}$. $\blacksquare$

This is the precise sense in which the solution set of an inhomogeneous system is not a set of the same kind as $N(A)$: it is a translate of $N(A)$, shifted away from the origin by any particular solution.
</details>
<br>


(d) Give an explicit $2 \times 2$ example with $\mathbf{b} \neq \mathbf{0}$ showing that the sum of two solutions need not be a solution.

<details>
    <br>
    <summary>
        View Solution
    </summary>

Take

$$
A = \begin{bmatrix} 1 & 0 \\ 0 & 0 \end{bmatrix}, \qquad \mathbf{b} = \begin{bmatrix} 1 \\ 0 \end{bmatrix}
$$

The vectors $\mathbf{x}_1 = (1,0)^T$ and $\mathbf{x}_2 = (1,5)^T$ both satisfy $A\mathbf{x} = \mathbf{b}$. Their sum is $(2,5)^T$, and

$$
A\begin{bmatrix} 2 \\ 5 \end{bmatrix} = \begin{bmatrix} 2 \\ 0 \end{bmatrix} \neq \mathbf{b}
$$

so the sum is not a solution. $\blacksquare$
</details>

</H2>

__Topic: Solution Sets.__

Let $A$ be $m \times n$ over $F$ and $\mathbf{b} \in F^m$.

(a) Prove Theorem 4: for every $\mathbf{b} \in F^m$ the system $A\mathbf{x} = \mathbf{b}$ has at most one solution if and only if $N(A) = \{\mathbf{0}\}$. Then explain why the quantifier over $\mathbf{b}$ cannot be dropped.

<details>
    <br>
    <summary>
        View Solution
    </summary>

__Proof.__ $(\Leftarrow)$ Suppose $N(A) = \{\mathbf{0}\}$. Fix $\mathbf{b} \in F^m$ and let $\mathbf{x}_1, \mathbf{x}_2$ both satisfy $A\mathbf{x} = \mathbf{b}$. Then

$$
A(\mathbf{x}_1 - \mathbf{x}_2) = A\mathbf{x}_1 - A\mathbf{x}_2 = \mathbf{b} - \mathbf{b} = \mathbf{0}
$$

so $\mathbf{x}_1 - \mathbf{x}_2 \in N(A) = \{\mathbf{0}\}$, giving $\mathbf{x}_1 = \mathbf{x}_2$. Hence for that $\mathbf{b}$, and therefore for every $\mathbf{b}$, there is at most one solution.

$(\Rightarrow)$ We prove the contrapositive. Suppose $N(A) \neq \{\mathbf{0}\}$ and choose $\mathbf{z} \in N(A)$ with $\mathbf{z} \neq \mathbf{0}$. Take $\mathbf{b} = \mathbf{0}$. Both $\mathbf{0}$ and $\mathbf{z}$ solve $A\mathbf{x} = \mathbf{0}$, and they are distinct, so for this particular $\mathbf{b}$ the system has more than one solution. Hence it is not the case that every $\mathbf{b}$ gives at most one solution. $\blacksquare$

__Why the quantifier is needed.__ Without it the statement is false. Take

$$
A = \begin{bmatrix} 1 & 0 \\ 0 & 0 \end{bmatrix}, \qquad \mathbf{b} = \begin{bmatrix} 0 \\ 1 \end{bmatrix}
$$

This system is inconsistent, so it has no solutions at all, and "at most one solution" is true vacuously. Yet $\mathbf{z} = (0,1)^T$ is a nonzero element of $N(A)$, so $N(A) \neq \{\mathbf{0}\}$ and the biconditional fails for that single $\mathbf{b}$.

The alternative repair is to add consistency as a hypothesis: a __consistent__ system has exactly one solution if and only if $N(A) = \{\mathbf{0}\}$. Both versions are correct, and noticing that the naive version is not is the point of the exercise. A statement of the form "at most one" is always true when there are none, and vacuous truth is where informal reasoning about uniqueness usually goes wrong.
</details>
<br>


(b) Prove that if $n > m$, then $N(A) \neq \{\mathbf{0}\}$. You may use the fact from Chapter One that the reduced row echelon form of $A$ has at most one pivot per row.

<details>
    <br>
    <summary>
        View Solution
    </summary>

__Proof.__ Let $R$ be the reduced row echelon form of $A$. Each pivot occupies its own row, so the number of pivot columns is at most $m$. Since $A$ has $n > m$ columns, at least one column of $R$ is free.

Assign the value $1$ to one free variable and $0$ to all other free variables, then solve for the pivot variables by back substitution. This produces a vector $\mathbf{z}$ satisfying $R\mathbf{z} = \mathbf{0}$, and $\mathbf{z} \neq \mathbf{0}$ because the entry corresponding to the chosen free variable equals $1$.

Row operations do not change the solution set of a homogeneous system, since each operation is reversible and each is an equation-preserving manipulation. Hence $A\mathbf{z} = \mathbf{0}$ as well, so $\mathbf{z} \in N(A)$ and $N(A) \neq \{\mathbf{0}\}$. $\blacksquare$
</details>
<br>


(c) Combine (a) and (b) into a single statement about underdetermined systems, and explain in one sentence what it means for a data set with more features than observations.

<details>
    <br>
    <summary>
        View Solution
    </summary>

__Statement.__ If $n > m$, then $A\mathbf{x} = \mathbf{b}$ never has exactly one solution: it has either no solutions or infinitely many.

For a design matrix with more features than observations, this says the coefficients can never be determined by the data alone. Either the model fails to fit at all, or it fits in infinitely many ways, and choosing among them requires information from outside the linear system, which in practice means a regularization criterion or additional constraints.
</details>

__Topic: Counterexamples.__

For each statement, decide whether it is true or false. If true, give a proof. If false, give an explicit counterexample and verify that it fails. Throughout, $A$ is a real matrix and vectors have real entries.

(a) If $A\mathbf{x} = \mathbf{b}$ is consistent for some $\mathbf{b} \neq \mathbf{0}$, then it is consistent for every $\mathbf{b}$.

<details>
    <br>
    <summary>
        View Solution
    </summary>

__False.__ Take

$$
A = \begin{bmatrix} 1 & 0 \\ 0 & 0 \end{bmatrix}
$$

With $\mathbf{b} = (1,0)^T$ the vector $\mathbf{x} = (1,0)^T$ is a solution, so the system is consistent. With $\mathbf{b} = (0,1)^T$ the second equation reads $0 = 1$, which is impossible, so the system is inconsistent. $\blacksquare$
</details>
<br>


(b) If $A$ has more rows than columns, then $A\mathbf{x} = \mathbf{b}$ is inconsistent.

<details>
    <br>
    <summary>
        View Solution
    </summary>

__False.__ Take

$$
A = \begin{bmatrix} 1 \\ 1 \end{bmatrix}, \qquad \mathbf{b} = \begin{bmatrix} 2 \\ 2 \end{bmatrix}
$$

Then $A$ is $2 \times 1$, with more rows than columns, and $\mathbf{x} = (2)$ satisfies $A\mathbf{x} = \mathbf{b}$. The system is consistent. $\blacksquare$

Overdetermined systems are usually inconsistent, and "usually" is not "always"; the distinction is exactly what a counterexample is for.
</details>
<br>


(c) If $A$ has more columns than rows, then $A\mathbf{x} = \mathbf{0}$ has a nonzero solution.

<details>
    <br>
    <summary>
        View Solution
    </summary>

__True.__ This is Exercise 7(b), and the proof given there applies: the reduced row echelon form has at most one pivot per row, hence at most $m$ pivots among $n > m$ columns, so some column is free. Setting that free variable to $1$ and the others to $0$ and back substituting produces a nonzero solution of the homogeneous system. $\blacksquare$
</details>
<br>


(d) The set $\{\mathbf{x} \in \mathbb{R}^3 : x_1 + x_2 + x_3 = 0\}$ is closed under addition and scalar multiplication.

<details>
    <br>
    <summary>
        View Solution
    </summary>

__True.__ __Proof.__ Let $\mathbf{u}, \mathbf{v}$ belong to the set and let $c \in \mathbb{R}$. Then $u_1 + u_2 + u_3 = 0$ and $v_1 + v_2 + v_3 = 0$. The entries of $\mathbf{u} + \mathbf{v}$ sum to

$$
(u_1+v_1) + (u_2+v_2) + (u_3+v_3) = (u_1+u_2+u_3) + (v_1+v_2+v_3) = 0 + 0 = 0
$$

using commutativity and associativity of addition in $\mathbb{R}$, so $\mathbf{u} + \mathbf{v}$ belongs to the set. Likewise the entries of $c\mathbf{u}$ sum to $cu_1 + cu_2 + cu_3 = c(u_1+u_2+u_3) = c \cdot 0 = 0$ by distributivity, so $c\mathbf{u}$ belongs to the set. $\blacksquare$

Observe that the set is $N(A)$ for the $1\times 3$ matrix $A = \begin{bmatrix} 1 & 1 & 1\end{bmatrix}$, so Theorem 2 gives the result immediately as well.
</details>
<br>


(e) The set $\{\mathbf{x} \in \mathbb{R}^3 : x_1 x_2 = 0\}$ is closed under addition.

<details>
    <br>
    <summary>
        View Solution
    </summary>

__False.__ Take $\mathbf{u} = (1,0,0)^T$ and $\mathbf{v} = (0,1,0)^T$. Both belong to the set, since $1 \cdot 0 = 0$ and $0 \cdot 1 = 0$. Their sum is $(1,1,0)^T$, whose first two entries have product $1 \cdot 1 = 1 \neq 0$, so the sum does not belong to the set. $\blacksquare$

The defining condition is not a linear equation, which is what allows the failure. Note that the set is a union of two planes, and the counterexample is the same phenomenon as Exercise 5(d).
</details>
<br>


(f) If $\mathbf{z} \in N(A)$ and $\mathbf{z} \neq \mathbf{0}$, then $A\mathbf{x} = \mathbf{b}$ has infinitely many solutions.

<details>
    <br>
    <summary>
        View Solution
    </summary>

__False.__ The system may have none. Take

$$
A = \begin{bmatrix} 1 & 0 \\ 0 & 0 \end{bmatrix}, \qquad \mathbf{b} = \begin{bmatrix} 0 \\ 1 \end{bmatrix}
$$

Then $\mathbf{z} = (0,1)^T$ is a nonzero element of $N(A)$, yet the second equation reads $0 = 1$ and the system has no solutions at all. $\blacksquare$

The correct statement adds a hypothesis: if the system is __consistent__ and $N(A) \neq \{\mathbf{0}\}$, then it has infinitely many solutions, by Theorem 3. Locating the missing hypothesis is the point of the exercise.
</details>
<br>

</H2>

__Topic: Matrix Algebra.__

In each part, prove the identity by comparing $i$th entries, and name the field axiom or definition used at every step. Let $A$ and $B$ be $m \times n$ over $F$, let $\mathbf{u}, \mathbf{v} \in F^n$, and let $c, d \in F$.

(a) $(A + B)\mathbf{u} = A\mathbf{u} + B\mathbf{u}$

<details>
    <br>
    <summary>
        View Solution
    </summary>

__Proof.__ Fix $i$ with $1 \leq i \leq m$.

$$
\begin{align*}
    \left[(A+B)\mathbf{u}\right]_i
        &= \sum_{j=1}^{n} (a_{ij} + b_{ij})u_j && \text{definitions of matrix sum and product} \\
        &= \sum_{j=1}^{n} \left(a_{ij}u_j + b_{ij}u_j\right) && \text{distributivity in } F \\
        &= \sum_{j=1}^{n} a_{ij}u_j + \sum_{j=1}^{n} b_{ij}u_j && \text{associativity and commutativity of addition} \\
        &= \left[A\mathbf{u}\right]_i + \left[B\mathbf{u}\right]_i && \text{definition of the product}
\end{align*}
$$

Since $i$ was arbitrary, the vectors are equal. $\blacksquare$
</details>
<br>


(b) $c(d\mathbf{u}) = (cd)\mathbf{u}$

<details>
    <br>
    <summary>
        View Solution
    </summary>

__Proof.__ Fix $i$ with $1 \leq i \leq n$. The $i$th entry of $c(d\mathbf{u})$ is $c(du_i)$ by the definition of scalar multiplication applied twice. By associativity of multiplication in $F$, $c(du_i) = (cd)u_i$, which is the $i$th entry of $(cd)\mathbf{u}$. Since $i$ was arbitrary the vectors are equal. $\blacksquare$

Only one axiom is used, and identifying it is the exercise.
</details>
<br>


(c) In the proof of Theorem 5 the step $\sum a_{ij}(cu_j) = \sum c(a_{ij}u_j)$ used commutativity of multiplication. Show precisely where, by writing that single step with each factor's position tracked, and state what the identity $A(c\mathbf{u}) = c(A\mathbf{u})$ would become if scalars were required to multiply from the right and did not commute.

<details>
    <br>
    <summary>
        View Solution
    </summary>

Working with one term of the sum, the entry of $A(c\mathbf{u})$ is built as $a_{ij} \cdot (c \cdot u_j)$. Associativity regroups this as $(a_{ij} \cdot c) \cdot u_j$, and to reach $c \cdot (a_{ij} \cdot u_j)$ one must replace $a_{ij}c$ by $c\,a_{ij}$. That replacement is the commutative law and nothing else; associativity alone does not permit it, since associativity may regroup factors but may not reorder them.

If scalars do not commute, the product $a_{ij}c$ cannot be rewritten as $c\,a_{ij}$, and the scalar cannot be pulled out to the left of the sum. What survives is a right-handed identity. Writing scalar multiplication on the right, so that the $j$th entry of $\mathbf{u}c$ is $u_j c$, the computation

$$
\left[A(\mathbf{u}c)\right]_i = \sum_{j=1}^n a_{ij}(u_j c) = \left(\sum_{j=1}^n a_{ij}u_j\right)c = \left[(A\mathbf{u})c\right]_i
$$

goes through using only associativity and distributivity. So over a system with noncommutative multiplication one obtains $A(\mathbf{u}c) = (A\mathbf{u})c$ and must abandon the left-handed version, which is why such a theory must fix a side for scalars at the outset.
</details>
<br>

__Topic: The Field $\mathbb{C}$.__

For $\mathbf{z} \in \mathbb{C}^n$, write $\bar{z}$ for the complex conjugate of $z$ and $\mathbf{z}^H = \bar{\mathbf{z}}^T$ for the conjugate transpose, so that $\mathbf{z}^H\mathbf{w} = \sum_{i=1}^n \bar{z}_i w_i$. Recall $|z|^2 = z\bar{z}$ for $z \in \mathbb{C}$.

(a) Prove that $\mathbf{z}^H\mathbf{z}$ is a nonnegative real number for every $\mathbf{z} \in \mathbb{C}^n$, and that it equals $0$ if and only if $\mathbf{z} = \mathbf{0}$.

<details>
    <br>
    <summary>
        View Solution
    </summary>

__Proof.__ By definition,

$$
\mathbf{z}^H\mathbf{z} = \sum_{i=1}^{n} \bar{z}_i z_i = \sum_{i=1}^{n} |z_i|^2
$$

Each $|z_i|$ is a nonnegative real number, hence each $|z_i|^2$ is a nonnegative real number, and a finite sum of nonnegative reals is a nonnegative real. Therefore $\mathbf{z}^H\mathbf{z} \geq 0$.

If $\mathbf{z} = \mathbf{0}$ then every $z_i = 0$ and the sum is $0$.

Conversely, suppose $\mathbf{z}^H\mathbf{z} = 0$. Each term $|z_i|^2$ is nonnegative, so if some $|z_k|^2 > 0$ then the total would be at least $|z_k|^2 > 0$, contradicting the hypothesis. Hence $|z_i|^2 = 0$ for every $i$, so $|z_i| = 0$ and $z_i = 0$ for every $i$, giving $\mathbf{z} = \mathbf{0}$. $\blacksquare$
</details>
<br>


(b) Exhibit a nonzero $\mathbf{z} \in \mathbb{C}^2$ with $\mathbf{z}^T\mathbf{z} = 0$, and verify it.

<details>
    <br>
    <summary>
        View Solution
    </summary>

Take $\mathbf{z} = (1,\ i)^T$, which is nonzero. Then

$$
\mathbf{z}^T\mathbf{z} = 1 \cdot 1 + i \cdot i = 1 + i^2 = 1 - 1 = 0
$$

while, by contrast, $\mathbf{z}^H\mathbf{z} = |1|^2 + |i|^2 = 1 + 1 = 2$. $\blacksquare$
</details>
<br>


(c) Explain why part (b) means that $\sqrt{\mathbf{z}^T\mathbf{z}}$ cannot serve as a measure of the size of a complex vector, and why the conjugate is not an arbitrary convention. Relate your answer to the fact from 1.P4.1 that $\mathbb{C}$ is not an ordered field.

<details>
    <br>
    <summary>
        View Solution
    </summary>

Any reasonable measure of size must vanish only at the zero vector; otherwise a nonzero vector would be indistinguishable from $\mathbf{0}$, and one could not say that an approximation with small error is close to its target. Part (b) exhibits a nonzero vector with $\mathbf{z}^T\mathbf{z} = 0$, so that quantity fails the requirement. It can also be negative or non-real for other choices of $\mathbf{z}$, so even its square root need not be defined.

The conjugate is what forces each term to be $z\bar{z} = |z|^2$, a nonnegative real, and part (a) shows this repairs both defects at once.

The connection to ordering is that inequalities such as $\mathbf{z}^H\mathbf{z} \geq 0$ are meaningless for a general complex number, since $\mathbb{C}$ admits no order compatible with its arithmetic. Part (a) is doing real work precisely because it proves the quantity is real before comparing it with $0$. Every statement about size, error, or convergence for complex vectors routes through a real-valued quantity for this reason, which is why complex-valued models in signal processing and machine learning are trained against real-valued loss functions built from $|z|^2$.
</details>
<br>

__Topic: Applications.__

This exercise proves, as a theorem, the phenomenon you observed computationally in Exercise 2 of Problem Set 1.P3: a design matrix with a redundant feature cannot have identifiable coefficients. Let $X$ be an $m \times n$ real matrix with columns $\mathbf{x}_1, \ldots, \mathbf{x}_n$, and let $\mathbf{y} \in \mathbb{R}^m$.

(a) Suppose columns $j$ and $k$ of $X$ are equal, with $j \neq k$. Prove that $N(X) \neq \{\mathbf{0}\}$ by exhibiting a nonzero element.

<details>
    <br>
    <summary>
        View Solution
    </summary>

__Proof.__ Let $\mathbf{z}$ be the vector whose $j$th entry is $1$, whose $k$th entry is $-1$, and whose remaining entries are $0$. Then $\mathbf{z} \neq \mathbf{0}$ because its $j$th entry is $1 \neq 0$. By the column form of the matrix-vector product,

$$
X\mathbf{z} = 1 \cdot \mathbf{x}_j + (-1) \cdot \mathbf{x}_k = \mathbf{x}_j - \mathbf{x}_k = \mathbf{0}
$$

since $\mathbf{x}_j = \mathbf{x}_k$ by hypothesis. Hence $\mathbf{z} \in N(X)$ and $N(X) \neq \{\mathbf{0}\}$. $\blacksquare$
</details>
<br>


(b) Conclude that if $X\boldsymbol{\beta} = \mathbf{y}$ has a solution, it has infinitely many, and that all of them give the same predictions $X\boldsymbol{\beta}$.

<details>
    <br>
    <summary>
        View Solution
    </summary>

__Proof.__ Suppose $\boldsymbol{\beta}_0$ is a solution. By Theorem 3 the solution set is $\{\boldsymbol{\beta}_0 + \mathbf{z} : \mathbf{z} \in N(X)\}$. By part (a) there is a nonzero $\mathbf{z} \in N(X)$, and by Theorem 2 every scalar multiple $t\mathbf{z}$ also lies in $N(X)$. The vectors $\boldsymbol{\beta}_0 + t\mathbf{z}$ are distinct for distinct real $t$, since $\mathbf{z} \neq \mathbf{0}$, so there are infinitely many solutions.

Every solution produces the same predictions: if $X\boldsymbol{\beta}_1 = \mathbf{y}$ and $X\boldsymbol{\beta}_2 = \mathbf{y}$ then both prediction vectors equal $\mathbf{y}$. Equivalently, $X(\boldsymbol{\beta}_0 + t\mathbf{z}) = X\boldsymbol{\beta}_0 + tX\mathbf{z} = \mathbf{y} + \mathbf{0} = \mathbf{y}$ by Theorem 5. $\blacksquare$
</details>
<br>


(c) Generalize: prove that if some column of $X$ is a linear combination of the others, the same conclusion holds. State the general principle in one sentence.

<details>
    <br>
    <summary>
        View Solution
    </summary>

__Proof.__ Suppose $\mathbf{x}_k = \sum_{j \neq k} c_j \mathbf{x}_j$ for scalars $c_j$. Define $\mathbf{z}$ by $z_k = -1$ and $z_j = c_j$ for $j \neq k$. Then $\mathbf{z} \neq \mathbf{0}$ because $z_k = -1$, and

$$
X\mathbf{z} = \sum_{j \neq k} c_j \mathbf{x}_j + (-1)\mathbf{x}_k = \mathbf{x}_k - \mathbf{x}_k = \mathbf{0}
$$

so $\mathbf{z} \in N(X)$ and $N(X) \neq \{\mathbf{0}\}$. The argument of part (b) then applies verbatim. $\blacksquare$

__Principle.__ Coefficients are identifiable only when no column of the design matrix is a linear combination of the other columns; otherwise the data determines the predictions but not the coefficients.
</details>
<br>


(d) State the converse of the principle in (c) and decide whether it is true. That is: if no column is a linear combination of the others, must the coefficients be unique when a solution exists?

<details>
    <br>
    <summary>
        View Solution
    </summary>

__Converse.__ If no column of $X$ is a linear combination of the others, then a consistent system $X\boldsymbol{\beta} = \mathbf{y}$ has exactly one solution.

This is __true__. By Theorem 4 it suffices to show $N(X) = \{\mathbf{0}\}$. Suppose $\mathbf{z} \in N(X)$ with $\mathbf{z} \neq \mathbf{0}$, and let $k$ be an index with $z_k \neq 0$. From $\sum_{j} z_j \mathbf{x}_j = \mathbf{0}$ we may solve for $\mathbf{x}_k$, using that $z_k$ has a multiplicative inverse in $\mathbb{R}$:

$$
\mathbf{x}_k = \sum_{j \neq k} \left( -\frac{z_j}{z_k} \right) \mathbf{x}_j
$$

exhibiting column $k$ as a linear combination of the others, contrary to hypothesis. Hence $N(X) = \{\mathbf{0}\}$ and the solution is unique. $\blacksquare$

Note where the field axioms enter: dividing by $z_k$ requires the multiplicative inverse axiom, which is exactly the axiom the integers fail.
</details>
<br>

__Topic: Challenge.__

Let $\mathbb{Q}(\sqrt{2}) = \{\, a + b\sqrt{2} \,:\, a, b \in \mathbb{Q} \,\}$, a subset of $\mathbb{R}$. You may use without proof that $\sqrt{2}$, $\sqrt{3}$, and $\sqrt{6}$ are irrational.

(a) Prove that $\mathbb{Q}(\sqrt{2})$ is closed under addition and multiplication.

<details>
    <br>
    <summary>
        View Solution
    </summary>

__Proof.__ Let $a + b\sqrt{2}$ and $c + d\sqrt{2}$ belong to the set, with $a,b,c,d \in \mathbb{Q}$. Then

$$
(a + b\sqrt{2}) + (c + d\sqrt{2}) = (a+c) + (b+d)\sqrt{2}
$$

and $a + c$, $b + d$ are rational, so the sum belongs to the set. Also

$$
(a + b\sqrt{2})(c + d\sqrt{2}) = ac + ad\sqrt{2} + bc\sqrt{2} + bd \cdot 2 = (ac + 2bd) + (ad + bc)\sqrt{2}
$$

and $ac + 2bd$, $ad + bc$ are rational, so the product belongs to the set. $\blacksquare$
</details>
<br>


(b) Prove that if $a$ and $b$ are rational and not both zero, then $a^2 - 2b^2 \neq 0$.

<details>
    <br>
    <summary>
        View Solution
    </summary>

__Proof.__ Suppose $a^2 - 2b^2 = 0$ with $a, b$ rational, not both zero. If $b = 0$ then $a^2 = 0$, so $a = 0$, contradicting that they are not both zero. Hence $b \neq 0$, and

$$
\left( \frac{a}{b} \right)^2 = 2
$$

so $a/b$ is a rational number whose square is $2$. Then $|a/b| = \sqrt{2}$ would be rational, contradicting the irrationality of $\sqrt{2}$. Therefore $a^2 - 2b^2 \neq 0$. $\blacksquare$
</details>
<br>


(c) Prove that every nonzero element of $\mathbb{Q}(\sqrt{2})$ has a multiplicative inverse in $\mathbb{Q}(\sqrt{2})$, and conclude that $\mathbb{Q}(\sqrt{2})$ is a field.

<details>
    <br>
    <summary>
        View Solution
    </summary>

__Proof.__ Let $a + b\sqrt{2} \neq 0$ with $a, b$ rational. Then $a$ and $b$ are not both zero, so $a^2 - 2b^2 \neq 0$ by part (b), and we may compute

$$
\frac{1}{a + b\sqrt{2}} = \frac{a - b\sqrt{2}}{(a+b\sqrt{2})(a - b\sqrt{2})} = \frac{a - b\sqrt{2}}{a^2 - 2b^2} = \frac{a}{a^2-2b^2} + \frac{-b}{a^2-2b^2}\sqrt{2}
$$

Both coefficients are quotients of rationals with nonzero denominator, hence rational, so the inverse belongs to $\mathbb{Q}(\sqrt{2})$.

For the conclusion, $\mathbb{Q}(\sqrt{2})$ is a subset of $\mathbb{R}$ containing $0 = 0 + 0\sqrt{2}$ and $1 = 1 + 0\sqrt{2}$, closed under addition and multiplication by part (a), closed under negation since $-(a+b\sqrt{2}) = (-a) + (-b)\sqrt{2}$, and closed under inversion of nonzero elements by the above. The associative, commutative, and distributive laws hold because they hold in $\mathbb{R}$ and the operations are the same. Hence $\mathbb{Q}(\sqrt{2})$ satisfies every field axiom. $\blacksquare$
</details>
<br>


(d) Does $\mathbb{Q}(\sqrt{2})$ satisfy the completeness axiom? Justify your answer briefly.

<details>
    <br>
    <summary>
        View Solution
    </summary>

__No.__ First we show $\sqrt{3} \notin \mathbb{Q}(\sqrt{2})$. Suppose $\sqrt{3} = a + b\sqrt{2}$ with $a, b \in \mathbb{Q}$. Squaring,

$$
3 = a^2 + 2b^2 + 2ab\sqrt{2}
$$

so $2ab\sqrt{2} = 3 - a^2 - 2b^2$ is rational. Since $\sqrt{2}$ is irrational, this forces $ab = 0$.

If $b = 0$ then $\sqrt{3} = a$ is rational, contradicting the irrationality of $\sqrt{3}$.

If $a = 0$ then $\sqrt{3} = b\sqrt{2}$, and multiplying both sides by $\sqrt{2}$ gives $\sqrt{6} = 2b$, which is rational, contradicting the irrationality of $\sqrt{6}$.

Both cases are impossible, so $\sqrt{3} \notin \mathbb{Q}(\sqrt{2})$.

Now consider

$$
S = \left\{\, x \in \mathbb{Q}(\sqrt{2}) \,:\, x > 0 \text{ and } x^2 < 3 \,\right\}
$$

The set is nonempty, since $1 \in S$, and bounded above by $2$. Its least upper bound in $\mathbb{R}$ is $\sqrt{3}$, which does not lie in $\mathbb{Q}(\sqrt{2})$. An argument identical in structure to Exercise 2, with $3$ in place of $2$, shows that no element of $\mathbb{Q}(\sqrt{2})$ is a least upper bound for $S$: an element whose square is less than $3$ is not an upper bound, an element whose square exceeds $3$ is not least, and no element has square exactly $3$. Hence $\mathbb{Q}(\sqrt{2})$ fails the completeness axiom. $\blacksquare$

The moral: being a field is an algebraic property and being complete is not. Many fields sit between $\mathbb{Q}$ and $\mathbb{R}$, all of them failing completeness, and $\mathbb{R}$ is distinguished among ordered fields by completeness alone.
</details>
