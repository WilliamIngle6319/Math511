Consider the system below, where $a$ and $b$ are constants

$$
\begin{align*}
3x -\ y &= 3 \\
ax + by &= 6
\end{align*}
$$

(a) Find values for $a$ and $b$ for which the resulting system has a unique solution. Show that the solution is unique. Determine the solution.

(b) Find values for $a$ and $b$ for which the resulting system has infinitely many solutions. Show that linear system has infinitely many solutions. Determine the solution set.

(c) Find values for $a$ and $b$ for which the resulting system has no solution. Show that the linear system has no solution.


(d) Graph the lines for each of the linear systems in parts (a), (b), and (c) on separate graphs. Include your graphs on the same page with your answers to (a)-(c). Do not create your plots with pen and paper. You must create your plots using your favorite graphing software. Consider a linear system with three linear equations in $x$, $y$, and $z$. The graph of each linear equation is a plane in a 3-dimensional coordinate system.

In parts (a), (b), and (c) you are given a figure and a description of how the three planes meet. __Your task in each part is to write down a system of three linear equations, one equation for each plane, whose graph is the figure shown.__ You are not asked to solve the system, and you are not asked to reproduce the figure. Any system whose three planes meet in the manner described is a correct answer, and the simplest such system is the best answer. Equations as simple as $x = 0$ or $z = 4$ are expected.


(a) __Write a system of three linear equations whose graph is Figure 2.__ All three planes intersect in a single line, the $x$-axis. Thus there are infinitely many solutions to the linear system: all of the points on the common line. 

<img src="https://courses.math.wichita.edu/math511/ch1/p1/threeplainsA.svg" width="60%" title="Three Planes" alt="The plot of three intersecting planes in 3 dimensional space."> 
<center>Figure 2</center>


(b) __Write a system of three linear equations whose graph is Figure 3.__ The three planes intersect at a single point, the origin. Thus the linear system has exactly one solution: the point that lies on all three planes.

<img src="https://courses.math.wichita.edu/math511/ch1/p1/threeplainsB.svg" width="60%" title="Three Planes" alt="The plot of three intersecting planes in 3 dimensional space."> 
<center>Figure 3</center>

(c) __Write a system of three linear equations whose graph is Figure 4.__ Two of the planes are parallel and distinct. Yet the third plane is *not* parallel to them. Thus the pairwise intersection of the third plane with each parallel plane is a line, and these two lines are parallel to each other. Thus there are no solutions to the linear system because no point lies on all three planes simultaneously.

<img src="https://courses.math.wichita.edu/math511/ch1/p1/threeplainsC.svg" width="60%" title="Three Planes" alt="The plot of three intersecting planes in 3 dimensional space."> 
<center>Figure 4</center>

(d) Parts (a), (b), and (c) show three of the __eight__ distinct ways that the graphs of three planes can meet in three-dimensional space. Five ways remain.

For each of the five remaining cases, write down a simple system of three linear equations whose graph exhibits that case, describe in one sentence how the three planes meet, and state whether the system has no solution, exactly one solution, or infinitely many solutions.

Then graph each of your five systems and include the five graphs with your answers. Do not create your plots with pen and paper. You must create your plots using graphing software; I suggest <a href="https://www.mathworks.com/products/matlab.html" title="MathWorks" aria-label="Webpage for accessing MATLAB Online"><img src="https://www.mathworks.com/etc.clientlibs/mathworks/clientlibs/customer-ui/templates/common/resources/images/favicon.20240807145759133.ico" alt="MATLAB icon" style="width:21px;height:21px;"></a><a href="https://www.mathworks.com/products/matlab.html" title="MathWorks" aria-label="Webpage for accessing MATLAB Online"> MATLAB$\textregistered$</a>, <a href="https://geogebra.org/" aria-label="GeoGebra graphing and geometry website"><img src="https://courses.math.wichita.edu/resources/geogebra.ico" title="GeoGebra" alt="Mathematics Resources Website GeoGebra" style="width:21px;height:21px;"></a>[ GeoGebra$\textregistered$](https://geogebra.org/), or <a href="https://desmos.com" aria-label="Desmos online graphing calculator"><img src="https://courses.math.wichita.edu/resources/Desmos-image-L1.png" title="Desmos" alt="Geometry tool and online graphing calculator" style="width:21px;height:21px;"></a>[ Desmos$\textregistered$](https://desmos.com). Later chapters will require these tools, so this is a good place to become comfortable with plotting planes in one of them.

If you would like to see the eight cases before you start, many websites illustrate them; a search for the intersection of three planes will find them. As in parts (a) through (c), the simplest system that does the job is the best answer.
