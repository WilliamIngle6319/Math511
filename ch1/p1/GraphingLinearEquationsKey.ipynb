Consider the system below, where $a$ and $b$ are constants

$$
\begin{align*}
3x -\ y &= 3 \\
ax + by &= 6
\end{align*}
$$

(a) Find values for $a$ and $b$ for which the resulting system has a unique solution. Show that the solution is unique. Determine the solution.

<details>
    <br>
    <summary>
        View Solution
    </summary>
    Any choice of $a$ and $b$ that produces a line not parallel to $3x - y = 3$ will give a unique solution. The first equation has slope $3$, so we need a second equation with slope different from $3$. The simplest choice is $a = 1$, $b = 1$, giving       $$
    \begin{align*}
        3x - y &= 3 \\
        \ x + y &= 6
    \end{align*}
    $$
    Row-reduce the augmented matrix.
    $$
    \begin{align*}
        \left[ \begin{array}{cc|c}\ \ 3\ & -1\ &\ \ 3\ \\ \ \ 1\ &\ \ 1\ &\ \ 6\ \end{array} \right] \begin{array}{l} R_1\leftrightarrow R_2 \\ \\ \end{array}
        &\rightarrow \left[ \begin{array}{cc|c}\ \ 1\ &\ \ 1\ &\ \ 6\ \\ \ \ 3\ & -1\ &\ \ 3\ \end{array} \right] \begin{array}{l} \\ R_2 - 3R_1 \end{array} \\
        \\
        &\rightarrow \left[ \begin{array}{cc|c}\ \ 1\ &\ \ 1\ &\ \ 6\ \\ \ \ 0\ & -4\ & -15\ \end{array} \right] \begin{array}{l} \\ -\frac{1}{4}R_2 \end{array} \\
        \\
        &\rightarrow \left[ \begin{array}{cc|c}\ \ 1\ &\ \ 1\ &\ \ 6\ \\ \ \ 0\ &\ \ 1\ &\ \ \frac{15}{4}\ \end{array} \right] \begin{array}{l} R_1 - R_2 \\ \\ \end{array} \\
        \\
        &\rightarrow \left[ \begin{array}{cc|c}\ \ 1\ &\ \ 0\ &\ \ \frac{9}{4}\ \\ \ \ 0\ &\ \ 1\ &\ \ \frac{15}{4}\ \end{array} \right]
\end{align*}
    $$
    Both columns of the coefficient matrix are pivot columns, so the system is independent and consistent. The unique solution is $x = \frac{9}{4}$, $y = \frac{15}{4}$.
</details>
<br>

(b) Find values for $a$ and $b$ for which the resulting system has infinitely many solutions. Show that linear system has infinitely many solutions. Determine the solution set.

<details>
    <br>
    <summary>
        View Solution
    </summary>
    For infinitely many solutions, the second equation must represent the same line as the first equation. The first equation, when scaled by $2$, becomes $6x - 2y = 6$. But the right-hand side of the second equation is fixed at $6$, so we need the left-hand side to be a scalar multiple of $3x - y$ such that the constant on the right is also $6$. Multiplying the first equation by $2$ gives $6x - 2y = 6$. Take $a = 6$, $b = -2$.
    $$\begin{align*}
        3x - y &= 3 \\ 6x - 2y &= 6
    \end{align*}$$
    Row-reduce.
    $$\begin{align*}
        \left[ \begin{array}{cc|c}\ \ 3\ & -1\ &\ 3\ \\ \ \ 6\ & -2\ &\ 6\ \end{array} \right] \begin{array}{c} \\ \\ R_2-R_1 \end{array} &\rightarrow \left[ \begin{array}{cc|c}\ \ 3\ & -1\ &\ 3\ \\ \ \ 0\ &\ \ 0\ &\ 0\ \end{array} \right]
    \end{align*}$$
    The second column is a free column, so the system is dependent. The bottom row reads $0 = 0$, so the system is consistent. Letting $y = s$ be a free parameter and back-substituting into $3x - y = 3$ gives $x = 1 + \frac{s}{3}$. The solution set is
    $$ \left\{\, \begin{bmatrix}\ 1 + \frac{s}{3}\ \\ \ s\ \end{bmatrix} \,:\, s \in \mathbb{R} \,\right\}. $$
</details>
<br>

(c) Find values for $a$ and $b$ for which the resulting system has no solution. Show that the linear system has no solution.

<details>
    <br>
    <summary>
        View Solution
    </summary>
    For no solution, the two equations must represent parallel distinct lines. The left-hand side of the second equation must be a scalar multiple of the first equation's left-hand side, but the right-hand sides must not match that scalar multiple. From part (b), $a = 6$ and $b = -2$ scales the first equation's left-hand side by $2$. Scaling the right-hand side by $2$ would give $6$, which matches — so to make the system inconsistent we keep $a = 6$, $b = -2$ but the right-hand side $6$ is forced by the problem statement. Instead, choose $a = 3$ and $b = -1$ so the second left-hand side equals the first.
    $$\begin{align*}
        3x - y &= 3 \\
        3x - y &= 6
    \end{align*}$$
    Row-reduce.
    $$\begin{align*}
        \left[ \begin{array}{cc|c}\ \ 3\ & -1\ &\ \ 3\ \\ \ \ 3\ & -1\ &\ \ 6\ \end{array} \right] \begin{array}{l} \\ R_2 - R_1 \end{array}
        &\rightarrow \left[ \begin{array}{cc|c}\ \ 3\ & -1\ &\ \ 3\ \\ \ \ 0\ &\ \ 0\ &\ \ 3\ \end{array} \right]
    \end{align*}$$
    The bottom row reads $0 = 3$, which is inconsistent. The linear system has no solution.
</details>
<br>

Consider a linear system with three linear equations in $x$, $y$, and $z$. The graph of each linear equation is a plane in a 3-dimensional coordinate system.

In parts (a), (b), and (c) you are given a figure and a description of how the three planes meet. __Your task in each part is to write down a system of three linear equations, one equation for each plane, whose graph is the figure shown.__ You are not asked to solve the system, and you are not asked to reproduce the figure. Any system whose three planes meet in the manner described is a correct answer, and the simplest such system is the best answer. Equations as simple as $x = 0$ or $z = 4$ are expected.


(a) __Write a system of three linear equations whose graph is Figure 2.__ All three planes intersect in a single line, the $x$-axis. Thus there are infinitely many solutions to the linear system: all of the points on the common line. 

<img src="https://courses.math.wichita.edu/math511/ch1/p1/threeplainsA.svg" width="60%" title="Three Planes" alt="The plot of three intersecting planes in 3 dimensional space."> 
<center>Figure 2</center>

<details>
    <br>
    <summary>
        View Solution
    </summary>
    The $x$-axis is the set of points $(x, 0, 0)$ for $x \in \mathbb{R}$, so it is the common intersection of all planes containing the $x$-axis. Any plane that contains the $x$-axis has an equation of the form $\beta y + \gamma z = 0$ for some $\beta$ and $\gamma$ not both zero. Three such planes that are pairwise distinct will intersect exactly along the $x$-axis. One example is
    $$ \begin{align*}
        z &= 0 \\
        y - z &= 0 \\
        y + z &= 0 
    \end{align*}$$
    The first equation is the $xy$-plane; the second is a plane through the $x$-axis tilted $45^{\circ}$ between the $xy$ and $xz$-planes; the third is a plane through the $x$-axis tilted $135^{\circ}$ between them. Each plane contains every point of the form $(x, 0, 0)$, so the solution set is
    $$\left\{\, \begin{bmatrix}\ s\ \\ \ 0\ \\ \ 0\ \end{bmatrix} \,:\, s \in \mathbb{R} \,\right\}.$$
</details>
<br>

(b) __Write a system of three linear equations whose graph is Figure 3.__ The three planes intersect at a single point, the origin. Thus the linear system has exactly one solution: the point that lies on all three planes.

<img src="https://courses.math.wichita.edu/math511/ch1/p1/threeplainsB.svg" width="60%" title="Three Planes" alt="The plot of three intersecting planes in 3 dimensional space."> 
<center>Figure 3</center>

<details>
    <br>
    <summary>
        View Solution
    </summary>
    Three planes meeting at exactly the origin must be mutually non-parallel and must share no line. The three coordinate planes themselves work:
    $$
    \begin{align*}
        x\ \ \ \ \ \ \ \ \ \ \ \ \ &= 0 \\
        \ \ \ \ \ \ \ y\ \ \ \ \ \ \ &= 0 \\
        \ \ \ \ \ \ \ \ \ \ \ \ \ z &= 0
    \end{align*}
    $$
    Row-reducing the augmented matrix gives the identity matrix with right-hand side $\mathbf{0}$. Both the coefficient matrix and the augmented matrix have three pivots, so the system is consistent and independent. The unique solution is $x=0$, $y=0$, $z=0$, the origin.
</details>
<br>

(c) __Write a system of three linear equations whose graph is Figure 4.__ Two of the planes are parallel and distinct. Yet the third plane is *not* parallel to them. Thus the pairwise intersection of the third plane with each parallel plane is a line, and these two lines are parallel to each other. Thus there are no solutions to the linear system because no point lies on all three planes simultaneously.

<img src="https://courses.math.wichita.edu/math511/ch1/p1/threeplainsC.svg" width="60%" title="Three Planes" alt="The plot of three intersecting planes in 3 dimensional space."> 
<center>Figure 4</center>

<details>
    <br>
    <summary>
        View Solution
    </summary>
    Two parallel distinct planes can be obtained by taking $z = 0$ and $z = 4$. A third plane that contains the $x$-axis (which lies in $z = 0$) but is not parallel to $z = 0$ will intersect $z = 0$ in the $x$-axis and $z = 4$ in a line parallel to the $x$-axis but distinct from it. Take $y = 0$ as the third plane.
    $$
    \begin{align*}
    \quad\quad\quad\quad z &= 0 \\
    \quad\quad\quad\quad z &= 4 \\
    \quad\quad y\quad\quad &= 0
    \end{align*}
    $$
    The first two equations together require $z$ to equal both $0$ and $4$, which is impossible. The system is inconsistent and has no solutions.
</details>
<br>

(d) Parts (a), (b), and (c) show three of the __eight__ distinct ways that the graphs of three planes can meet in three-dimensional space. Five ways remain.

For each of the five remaining cases, write down a simple system of three linear equations whose graph exhibits that case, describe in one sentence how the three planes meet, and state whether the system has no solution, exactly one solution, or infinitely many solutions.

Then graph each of your five systems and include the five graphs with your answers. Do not create your plots with pen and paper. You must create your plots using graphing software; I suggest <a href="https://www.mathworks.com/products/matlab.html" title="MathWorks" aria-label="Webpage for accessing MATLAB Online"><img src="https://www.mathworks.com/etc.clientlibs/mathworks/clientlibs/customer-ui/templates/common/resources/images/favicon.20240807145759133.ico" alt="MATLAB icon" style="width:21px;height:21px;"></a><a href="https://www.mathworks.com/products/matlab.html" title="MathWorks" aria-label="Webpage for accessing MATLAB Online"> MATLAB$\textregistered$</a>, <a href="https://geogebra.org/" aria-label="GeoGebra graphing and geometry website"><img src="https://courses.math.wichita.edu/resources/geogebra.ico" title="GeoGebra" alt="Mathematics Resources Website GeoGebra" style="width:21px;height:21px;"></a>[ GeoGebra$\textregistered$](https://geogebra.org/), or <a href="https://desmos.com" aria-label="Desmos online graphing calculator"><img src="https://courses.math.wichita.edu/resources/Desmos-image-L1.png" title="Desmos" alt="Geometry tool and online graphing calculator" style="width:21px;height:21px;"></a>[ Desmos$\textregistered$](https://desmos.com). Later chapters will require these tools, so this is a good place to become comfortable with plotting planes in one of them.

If you would like to see the eight cases before you start, many websites illustrate them; a search for the intersection of three planes will find them. As in parts (a) through (c), the simplest system that does the job is the best answer.

<details>
    <br>
    <summary>
        View Solution
    </summary>

Parts (a), (b), and (c) covered three cases: three planes meeting in a common line, three planes meeting at a single point, and two parallel distinct planes cut by a third. The five remaining cases follow. In each, the simplest equations available are used, and any equivalent answer is correct.

__1. All three planes coincide.__

$$
\begin{align*}
    \ \ x + y + z &= 0 \\
    2x + 2y + 2z &= 0 \\
    3x + 3y + 3z &= 0
\end{align*}
$$

The second and third equations are multiples of the first, so all three graphs are the same plane. Every point of that plane satisfies all three equations, so the system has infinitely many solutions, filling a plane.

__2. Two planes coincide and the third is parallel to them but distinct.__

$$
\begin{align*}
    \ z &= 0 \\
    2z &= 0 \\
    \ z &= 1
\end{align*}
$$

The first two equations describe the same plane and the third describes a plane parallel to it. No point lies on both, so the system has no solution.

__3. Two planes coincide and the third is not parallel to them.__

$$
\begin{align*}
    \ z &= 0 \\
    2z &= 0 \\
    \ y &= 0
\end{align*}
$$

The first two graphs are the same plane, and the third cuts it in a line, here the $x$-axis. The system has infinitely many solutions, filling a line.

__4. Three distinct parallel planes.__

$$
\begin{align*}
    z &= 0 \\
    z &= 1 \\
    z &= 2
\end{align*}
$$

No two of the planes share a point, so the system has no solution. Note that any two of these equations already contradict each other.

__5. Three planes meeting pairwise in three distinct parallel lines.__

$$
\begin{align*}
    x + y \ \ \ &= 0 \\
    \ \ \ y + z &= 0 \\
    x \ \ \ - z &= 1
\end{align*}
$$

No two of these planes are parallel, so each pair meets in a line, giving three lines. Subtracting the second equation from the first gives $x - z = 0$, while the third equation requires $x - z = 1$, so no point lies on all three planes and the system has no solution. The three lines are parallel to one another and the figure resembles the surface of a triangular prism, which is the name usually given to this case.

<br>

Counting the cases by their solution sets: one case has exactly one solution, three cases have infinitely many, and four cases have none. Only the first of these can occur when the coefficient matrix is nonsingular, which is the connection between this exercise and the row reduction of Section 1.2.
</details>
