<H2><img src="https://courses.math.wichita.edu/resources/wu_athletics_32x32.png" title="Exercise - Sensor Calibration" alt="Exercise - linear data with noise" style="width:21px;height:21px;" class="anchor" id="Exercise:SensorCalibration"> Exercise 1: Sensor Calibration <img src="https://courses.math.wichita.edu/resources/wsu_logo.svg" align="right" alt="Wichita State University Logo" title="WSU logo" width="15%"></a>
</H2>
</div>

A K-type thermocouple produces a small voltage that depends approximately linearly on temperature. To calibrate the sensor, an engineer immerses the thermocouple in three temperature baths whose true temperatures are known (measured with a reference instrument). She records the voltage at each bath:

<table style="width:50%">
    <caption>Calibration Data</caption>
    <tr>
        <th scope="col">True Temperature</th>
        <th scope="col">Measured Voltage</th>
    </tr>
    <tr>
        <th scope="col">$T(^{\circ}C)$</th>
        <th scope="col">$V\,(mV)$</th>
    </tr>
    <tr>
        <td align='center'>25</td>
        <td align='center'>1.0</td>
    </tr>
    <tr>
        <td align='center'>25</td>
        <td align='center'>2.1</td>
    </tr>
    <tr>
        <td align='center'>100</td>
        <td align='center'>4.0</td>
    </tr>
</table>

She models the relationship as 

$$
V = aT + b
$$

where $a(mV/^{\circ}C)$ is the sensors' sensitivity and $b(mV)$ is its offset at $0^{\circ}C$.

(a) Write down the linear system for $a$ and $b$.

(b) What is the size of this linear system? Is it square, overdetermined, or undetermined?

(c) Use the first two equations to solve for $a$ and $b$. Then check whether the resulting line $V = aT+b$ correctly predicts the third measurement at $T=100^{\circ}C$? What do you conclude?

(d) Repeat part (c) using the last two equations to solve for $b$ and $c$. The check the prediction at $T=25^{\circ}C$. Do you get the same line?

(e) Explain in your own words why this linear system has no solution. What is physically happening in the calibration data?

<div class="alert alert-block alert-success">

This is a typical situation in engineering and data science: more measurements than unknowns, and no exact solution. Later in the course we will study least squares, a technique that finds the line $V = aT + b$ that comes closest to all three measurements — the line that minimizes the total prediction error even though no line satisfies all the data exactly.

</div>

<details>
    <br>
    <summary>
        View Solution
    </summary>
    <ol>
        <p>(a) Substituting each $(T,V)$ pair into $V = aT + b$:</p>
        $$ \begin{align*}
            25a + b &= 1.0 \\
            50a + b &= 2.1 \\
            100a + b &= 4.0
        \end{align*} $$
        <p>(b) The system has 3 equations and 2 unknowns, a $3\times 2$ linear system. Since there are more equations than unknowns, the linear system is overdetermined.</p>
        <p>(c) Subtracting the first equation from the second eliminates $b$.</p>
        $$ 25a = 1.1\ \Rightarrow\ a = 0.044 \textrm{ mV/}^\circ\textrm{C} $$
        <p>Substituting $a = 0.044$ into the first equation yields $b = 1.0 - 25(0.044) = -0.10$ mV. The line determined by the first two measurements is</p>
        $$ V = 0.044\,T - 0.10. $$
        <p>Check the third measurement at $T = 100$°C. The line predicts $V = 0.044(100) - 0.10 = 4.30$ mV, but the measured value is $4.0$ mV. The third equation is <b>not</b> satisfied.</p>
        <p>(d) Subtracting the first equation from the third eliminates $b$.</p>
        $$ 75a = 3.0\ \Rightarrow\ a = 0.040 \textrm{ mV/}^\circ\textrm{C} $$
        <p>Substituting $a = 0.040$ into the first equation yields $b = 1.0 - 25(0.040) = 0$ mV. The line determined by the first and third measurements is</p>
        $$ V = 0.040\,T. $$
        <p>Check the prediction at $T = 50$°C. The line predicts $V = 0.040(50) = 2.0$ mV, but the measured value is $2.1$ mV. The middle measurement is <b>not</b> satisfied either.</p>
        <p>The two pairs of equations produce <b>different lines</b>: $V = 0.044\,T - 0.10$ from one pair, $V = 0.040\,T$ from another. Neither line passes through all three data points.</p>
        <p>(e) The three measurements are not exactly collinear. Plotted on the $(T, V)$ plane, they do not lie on a single straight line. There is no choice of $a$ and $b$ that makes $V = aT + b$ pass through all three points simultaneously. The linear system has <b>no solutions</b>; it is inconsistent.</p>
        <p>Physically, the sensor's response is only approximately linear, and each measurement carries some noise from the bath temperature, the reference instrument, and the voltmeter. Real calibration data almost always produces an overdetermined, inconsistent system.</p>
    </ol>
    <br>
</details>

</H2>
</div>

A small electronics shop assembles three products from two limited resources each day: technician time (hours) and soldering-station time (hours). Each unit requires:
<table style="width:50%">
    <caption>Resource Requirements per Unit</caption>
    <thead>
        <tr>
            <th scope="col" style="text-align: center"></th>
            <th scope="col" style="text-align: center">Sensor module</th>
            <th scope="col" style="text-align: center">Controller board</th>
            <th scope="col" style="text-align: center">Display panel</th>
        </tr>
    </thead>
    <tbody>
        <tr>
            <th scope="row">Technician (hr)</th>
            <td style="text-align: center">2</td>
            <td style="text-align: center">3</td>
            <td style="text-align: center">1</td>
        <tr>
            <th scope="row">Soldering (hr)</th>
            <td style="text-align: center">1</td>
            <td style="text-align: center">1</td>
            <td style="text-align: center">2</td>
        </tr>
    </tbody>
</table>

Let $x_1$, $x_2$, and $x_3$ denote the number of sensor modules, controller boards, and display panels produced in a day. The shop has 24 technician-hours and 16 soldering-hours available each day, and the manager wants to use all of both resources (no idle staff, no idle equipment).

(a) Write down the linear system for $x_1$, $x_2$, and $x_3$.

(b) What is the size of this linear system? Is it square, overdetermined, or underdetermined?

(c) Eliminate $x_1$ between the two equations. Solve for $x_2$ in terms of $x_3$, and then for $x_1$ in terms of $x_3$. Describe the set of all solutions.

(d) A production plan is physically meaningful only if every $x_i \geq 0$, $1\le i\le3$. For which values of $x_3$ does the solution from part (c) satisfy $x_1 \ge 0$, $x_2 \ge 0$, and $x_3 \ge 0$? Give one physically meaningful example.

This is a typical situation in operations and economics: more variables than constraints produces a family of feasible plans, and the manager must impose additional considerations — profit, demand, inventory limits — to pick a single plan. Later in the course we will see techniques for representing and choosing among one-parameter families of solutions systematically.

<details>
    <br>
    <summary>
        View Solution
    </summary>
    <ol>
        <p>(a) Each row of the table gives one equation:</p>
        $$ \begin{align*}
            2x_1 + 3x_2 + x_3 &= 24 \\
            x_1 + x_2 + 2x_3 &= 16
        \end{align*} $$
        <p>(b) The system has $2$ equations and $3$ unknowns, a $2 \times 3$ linear system. Since there are fewer equations than unknowns, the linear system is underdetermined.</p>
        <p>(c) Multiply the second equation by $2$ to align the $x_1$ coefficient with the first equation.</p>
        $$ \begin{align*}
            2x_1 + 3x_2 + x_3 &= 24 \\
            2x_1 + 2x_2 + 4x_3 &= 32
        \end{align*} $$
        <p>Subtracting the second equation from the first eliminates $x_1$.</p>
        $$ x_2 - 3x_3 = -8 $$
        <p>Solve for $x_2$.</p>
        $$ x_2 = 3x_3 - 8 $$
        <p>Treat $x_3$ as a free parameter. Substituting back into the original second equation gives $x_1$.</p>
        $$ \begin{align*}
            x_1 + (3x_3 - 8) + 2x_3 &= 16 \\
            x_1 + 5x_3 - 8 &= 16 \\
            x_1 &= 24 - 5x_3
        \end{align*} $$
        <p>There are infinitely many solutions. The solution set is a <b>one-parameter family</b>:</p>
        $$ \left\{\, \left( 24 - 5x_3,\, 3x_3 - 8,\, x_3 \right) \,:\, x_3 \in \mathbb{R} \,\right\} $$
        <p>(d) The mathematical solution set is an entire line, but most points on that line are not physically meaningful. We need $x_1, x_2, x_3 \ge 0$ simultaneously.</p>
        <ul>
            <li>$x_3 \ge 0$ is given.</li>
            <li>$x_2 \ge 0$ requires $3x_3 - 8 \ge 0$, or $x_3 \ge \frac{8}{3}$.</li>
            <li>$x_1 \ge 0$ requires $24 - 5x_3 \ge 0$, or $x_3 \le \frac{24}{5}$.</li>
        </ul>
        <p>Hence $\frac{8}{3} \le x_3 \le \frac{24}{5}$, or approximately $2.67 \le x_3 \le 4.8$.</p>
        <p>One concrete example: take $x_3 = 4$ display panels.</p>
        $$ \begin{align*}
            x_2 &= 3(4) - 8 = 4 \textrm{ controller boards} \\
            x_1 &= 24 - 5(4) = 4 \textrm{ sensor modules}
        \end{align*} $$
        <p>Check:</p>
        $$ \begin{align*}
            2(4) + 3(4) + 1(4) &= 8 + 12 + 4 = 24 \textrm{ technician-hours} \\
            1(4) + 1(4) + 2(4) &= 4 + 4 + 8 = 16 \textrm{ soldering-hours}
        \end{align*} $$
        <p>So 4 sensor modules, 4 controller boards, and 4 display panels is a feasible daily production plan that uses every available hour of both resources.</p>
    </ol>
    <br>
</details>

A data scientist at a university wants to predict a student's end-of-semester score from three features she can measure during the semester: weekly hours studied, attendance rate, and prior GPA. She proposes the linear model

$$
s = a\cdot h + b\cdot r + c\cdot g
$$

where $s$ is the predicted score, $h$ is weekly hours studied, $r$ is attendance rate (between $0$ and $1$), $g$ is prior GPA (between $0$ and $4$), and $a$, $b$, $c$ are the coefficients she needs to learn from data.

She pulls records for three past students whose final scores are known:
<table style="width:60%"> <caption>Training Data</caption> <thead> <tr> <th scope="col" style="text-align: center">Student</th> <th scope="col" style="text-align: center">Hours studied $h$</th> <th scope="col" style="text-align: center">Attendance $r$</th> <th scope="col" style="text-align: center">Prior GPA $g$</th> <th scope="col" style="text-align: center">Score $s$</th> </tr> </thead> <tbody> <tr> <th scope="row">A</th> <td style="text-align: center">10</td> <td style="text-align: center">0.8</td> <td style="text-align: center">3.0</td> <td style="text-align: center">82</td> </tr> <tr> <th scope="row">B</th> <td style="text-align: center">5</td> <td style="text-align: center">1.0</td> <td style="text-align: center">2.0</td> <td style="text-align: center">61</td> </tr> <tr> <th scope="row">C</th> <td style="text-align: center">15</td> <td style="text-align: center">0.5</td> <td style="text-align: center">4.0</td> <td style="text-align: center">98</td> </tr> </tbody> </table>

(a) Write down the linear system for $a$, $b$, and $c$.

(b) What is the size of this linear system? Is it square, overdetermined, or underdetermined?

(c) Solve the linear system using Gaussian elimination.

(d) Use the fitted model to predict the end-of-semester score of a new student who studies $8$ hours per week, has an attendance rate of $0.9$, and a prior GPA of $3.5$.

<div class="alert alert-block alert-success">

This is what "fitting a linear model" literally means in data science and machine learning. Each row of the data table contributes one equation, each feature contributes one unknown coefficient, and solving the linear system produces the model. In a real application, the data scientist would use far more than three students, producing an overdetermined system that is solved approximately by least squares — a technique we will study later. The mathematics is the same; only the number of rows changes.

Notice that the fitted coefficient on prior GPA is negative. This is a warning about fitting a model to very little data: with only three students, any pattern the system happens to capture is treated as truth, even if it contradicts what we would expect physically. Real-world data scientists use far more data and additional techniques to avoid this.

</div>

    (a) Substituting each student's features and known score into the model $s = ah + br + cg$ produces one equation per student:
    $$ 
    \begin{align*} 10a + 0.8b + 3.0c &= 82 \\ 
        5a + 1.0b + 2.0c &= 61 \\ 
        15a + 0.5b + 4.0c &= 98 
    \end{align*} 
    $$
    <br>
    (b) The system has $3$ equations and $3$ unknowns, a $3 \times 3$ linear system. It is square.<br>
    (c) Eliminate $a$ from the second and third equations. Subtract one-half of the first equation from the second. 
    $$ 0.6b + 0.5c = 20 $$ 
    Subtract three-halves of the first equation from the third.
    $$ -0.7b - 0.5c = -25 $$ 
    The reduced system in $b$ and $c$ is 
    $$ \begin{align*} 0.6b + 0.5c &= 20 \\ -0.7b - 0.5c &= -25 \end{align*} $$
    Adding the two equations eliminates $c$.
    $$ -0.1b = -5\ \Rightarrow\ b = 50 $$
    Substituting $b = 50$ into $0.6b + 0.5c = 20$ gives
    $$ 30 + 0.5c = 20\ \Rightarrow\ c = -20 $$
    Substituting $b = 50$ and $c = -20$ into the first original equation gives
    $$
    \begin{align*}
        10a + 0.8(50) + 3.0(-20) &= 82 \\ 
        10a + 40 - 60 &= 82 \\ 
        10a &= 102 \\ 
        a &= 10.2 
    \end{align*}
    $$
    The fitted model is 
    $$ s = 10.2\,h + 50\,r - 20\,g. $$ 
    (d) Substitute the new student's features $h = 8$, $r = 0.9$, $g = 3.5$.</p> $$ s = 10.2(8) + 50(0.9) - 20(3.5) = 81.6 + 45 - 70 = 56.6 $$
    he model predicts a score of approximately $57$.
</details>

### Short Answer

#### 1. Explain why you would expect an overdetermined linear system to be inconsistent. 

<details> 
    <summary> View Solution </summary> 
    An overdetermined system has more equations than unknowns. Each equation imposes a constraint, and the unknowns must satisfy all of them simultaneously. With more constraints than degrees of freedom, the constraints will typically conflict — there is no choice of unknowns that satisfies every equation at once. An overdetermined system is consistent only when the extra equations happen to be redundant with (linear combinations of) the others, which is a special, non-generic situation.
</details>

#### 2.  Explain why you would expect an underdetermined consistent linear system to have infinitely many solutions rather than a unique one.

<details> 
    <summary> View Solution </summary> 
    Expected answer. An underdetermined system has fewer equations than unknowns. Each equation removes one degree of freedom from the unknowns. With fewer equations than unknowns, at least one degree of freedom remains uncontrolled after all equations are satisfied — a free variable that can take any value while the remaining unknowns adjust to keep every equation satisfied. The result is a one-parameter (or higher) family of solutions, not a single point.
</details>

#### 3. A student measures the temperature in a chemistry lab using three different thermometers placed in the same beaker of water. 

The student wants to find the "true" temperature by treating each reading as an equation $T = m_i$, where $m_i$ is the reading of thermometer $i$. Is the linear system square, overdetermined, or underdetermined? Why is it almost certainly inconsistent? What does this tell one about a plan to find a single "true" temperature?

<details> 
    <summary> View Solution </summary> 
    The system has three equations and one unknown $T$, so it is overdetermined. Real thermometers carry measurement error, so the three readings will almost never be exactly equal, and there is no value of $T$ that satisfies all three equations simultaneously — the system is inconsistent. Any plan to find a single value of $T$ that exactly matches every reading is unrealistic; instead, one needs a way of choosing a $T$ that comes close to all three readings (for example, by averaging, or by minimizing the total error).
</details>

#### 4. A traffic engineer monitors flow through a four-way intersection. 

The engineer writes down a conservation equation at each of the four corners (flow in equals flow out). The resulting system has four equations and several unknown flow rates on the connecting streets. Surprisingly, even though there are four equations, the system is still underdetermined. Give a plausible explanation for why the four conservation equations do not act as four independent constraints.

<details> 
    <summary> View Solution </summary> 
    In a closed network, the conservation equations at the nodes are not all independent. Whatever flows in to one corner must eventually leave through the others, so summing the equations at all four corners produces an identity ($0 = 0$) — the total flow in equals the total flow out by construction. This means one of the four equations is a linear combination of the other three; it is redundant. Effectively, only three of the four equations impose real constraints. If the network has more unknown flow rates than three, the system remains underdetermined and admits a family of feasible flow patterns.
</details>
